(b) Provide an estimate of the standard error of µˆ. Interpret this
result.

Hint: We can compute the standard error of the sample mean by
dividing the sample standard deviation by the square root of the
number of observations.

```python
s = np.std(boston['medv'], ddof=1)   # sample standard deviation
n = len(boston['medv'])
se = s / np.sqrt(n)
print(se)
```

(c) Now estimate the standard error of µˆ using the bootstrap. How does this compare to your answer from (b)?

```python
B = 1000
boot_means = np.empty(B)

for b in range(B):
    sample = boston.sample(n=n, replace=True)
    boot_means[b] = sample['medv'].mean()

boot_se = boot_means.std()
print(boot_se)
```

The bootstrap estimate is very close to the formula-based estimate.

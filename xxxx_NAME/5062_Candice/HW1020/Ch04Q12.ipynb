###12. Suppose that you wish to classify an observation X ∈ R into apples and oranges. You fit a logistic regression model and find that
$$ \hat{P}(Y = orange|X = x) = \frac{exp(\hat{\beta_0} + \hat{\beta_1x)}}{1 + exp(\hat{\beta_0} + \hat{\beta_1x})} $$


###Your friend fits a logistic regression model to the same data using thesoftmax formulation in (4.13), and finds that
$$
\hat{P}(Y = \text{orange} \mid X = x) =
\frac{\exp(\hat{\alpha}_{\text{orange},0} + \hat{\alpha}_{\text{orange},1} x)}
{\exp(\hat{\alpha}_{\text{orange},0} + \hat{\alpha}_{\text{orange},1} x) + \exp(\hat{\alpha}_{\text{apple},0} + \hat{\alpha}_{\text{apple},1} x)}.
$$

####(a) What is the log odds of orange versus apple in your model?

Log-odds:
$$
\log\left(\frac{p}{1 - p}\right)
$$

In the logistic regression model,  
$$
p = \widehat{P}(Y = \text{orange} \mid X = x) =
\frac{\exp(\widehat{\beta}_0 + \widehat{\beta}_1 x)}
{1 + \exp(\widehat{\beta}_0 + \widehat{\beta}_1 x)}.
$$

Then, the probability of apple is  
$$
\widehat{P}(Y = \text{apple} \mid X = x)
= 1 - \widehat{P}(Y = \text{orange} \mid X = x)
= \frac{1}{1 + \exp(\widehat{\beta}_0 + \widehat{\beta}_1 x)}.
$$

Therefore, the **log-odds** of orange versus apple is  
$$
\log\left(
\frac{\widehat{P}(Y = \text{orange} \mid X = x)}
{\widehat{P}(Y = \text{apple} \mid X = x)}
\right)
= \log\left( \exp(\widehat{\beta}_0 + \widehat{\beta}_1 x) \right)
= \widehat{\beta}_0 + \widehat{\beta}_1 x.
$$


####(b) What is the log odds of orange versus apple in your friend’s model?

 friend’s softmax model is:

$$
\widehat{P}(Y = \text{orange} \mid X = x) =
\frac{\exp(\widehat{\alpha}_{\text{orange},0} + \widehat{\alpha}_{\text{orange},1}x)}
{\exp(\widehat{\alpha}_{\text{orange},0} + \widehat{\alpha}_{\text{orange},1}x) + \exp(\widehat{\alpha}_{\text{apple},0} + \widehat{\alpha}_{\text{apple},1}x)}.
$$

Similarly,

$$
\widehat{P}(Y = \text{apple} \mid X = x) =
\frac{\exp(\widehat{\alpha}_{\text{apple},0} + \widehat{\alpha}_{\text{apple},1}x)}
{\exp(\widehat{\alpha}_{\text{orange},0} + \widehat{\alpha}_{\text{orange},1}x) + \exp(\widehat{\alpha}_{\text{apple},0} + \widehat{\alpha}_{\text{apple},1}x)}.
$$


Then, the log-odds of orange versus apple is:

$$
\log\left(
\frac{\widehat{P}(Y = \text{orange} \mid X = x)}
{\widehat{P}(Y = \text{apple} \mid X = x)}
\right)
= \log\left(
\frac{\exp(\widehat{\alpha}_{\text{orange},0} + \widehat{\alpha}_{\text{orange},1}x)}
{\exp(\widehat{\alpha}_{\text{apple},0} + \widehat{\alpha}_{\text{apple},1}x)}
\right).
$$

Simplifying gives:

$$
\log\left(
\frac{\widehat{P}(Y = \text{orange})}{\widehat{P}(Y = \text{apple})}
\right)
= (\widehat{\alpha}_{\text{orange},0} - \widehat{\alpha}_{\text{apple},0})
+ (\widehat{\alpha}_{\text{orange},1} - \widehat{\alpha}_{\text{apple},1})x.
$$

####(c) Suppose that in your model, $\hat{\beta_0}=2$, $\hat{\beta_1}=-1$. What are the coefficient estimates in your friend’s model? Be as specific as possible.

For binary logistic regression, the softmax formulation is equivalent to the standard logistic regression.  

The log-odds:

$$
\hat{\beta}_0 + \hat{\beta}_1 x = (\widehat{\alpha}_{\text{orange},0} - \widehat{\alpha}_{\text{apple},0}) + (\widehat{\alpha}_{\text{orange},1} - \widehat{\alpha}_{\text{apple},1})x
$$


By convention, set apple as the baseline class:

$$
\widehat{\alpha}_{\text{apple},0} = 0, \quad \widehat{\alpha}_{\text{apple},1} = 0
$$

Then the orange coefficients are:

$$
\widehat{\alpha}_{\text{orange},0} = \hat{\beta}_0 = 2, \quad
\widehat{\alpha}_{\text{orange},1} = \hat{\beta}_1 = -1.
$$

####(d)Now suppose that you and your friend fit the same two models on a different data set. This time, your friend gets the coefficient estimates:

$$
\widehat{\alpha}_{\text{orange},0} = 1.2, \quad \widehat{\alpha}_{\text{orange},1} = -2, \quad
\widehat{\alpha}_{\text{apple},0} = 3, \quad \widehat{\alpha}_{\text{apple},1} = 0.6.
$$

What are the coefficient estimates in your model?



Friend’s softmax coefficients:

$$
\widehat{\alpha}_{\text{orange},0} = 1.2, \quad \widehat{\alpha}_{\text{orange},1} = -2, \quad
\widehat{\alpha}_{\text{apple},0} = 3, \quad \widehat{\alpha}_{\text{apple},1} = 0.6
$$

Relationship between logistic regression and softmax coefficients:

$$
\hat{\beta}_0 + \hat{\beta}_1 x = (\widehat{\alpha}_{\text{orange},0} - \widehat{\alpha}_{\text{apple},0}) + (\widehat{\alpha}_{\text{orange},1} - \widehat{\alpha}_{\text{apple},1}) x
$$

Compute:

$$
\hat{\beta}_0 = 1.2 - 3 = -1.8, \quad \hat{\beta}_1 = -2 - 0.6 = -2.6
$$

Answer:

$$
\hat{\beta}_0 = -1.8, \quad \hat{\beta}_1 = -2.6
$$

####(e) Finally, suppose you apply both models from (d) to a data set with 2,000 test observations. What fraction of the time do you expect the predicted class labels from your model to agree with those from your friend’s model? Explain your answer.

For binary classification, logistic regression and softmax models produce the same log-odds up to a constant shift, so predicted classes are identical. Therefore, for 2,000 test observations:

Fraction of agreement=100%

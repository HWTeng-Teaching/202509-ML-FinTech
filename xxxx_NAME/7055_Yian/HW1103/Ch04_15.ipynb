(d) Now create a new function, Power3(), that actually returns the</br>
result x**a as a Python object, rather than simply printing it</br>
to the screen. That is, if you store the value x ** a in an object</br>
called result within your function, then you can simply return return</br>
this result, using the following line: return result</br>
Note that the line above should be the last line in your function,</br>
and it should be indented 4 spaces.</br>

```python
def Power3(x, a):
  result = x**a
  return result

print('The answer to 10^3 is: ', Power3(10, 3))
print('The answer to 8^17 is: ', Power3(8, 17))
print('The answer to 131^3 is: ', Power3(131, 3))
```

(e) Now using the Power3() function, create a plot of f (x) = x2. </br>
The x-axis should display a range of integers from 1 to 10, and</br>
the y-axis should display x2. Label the axes appropriately, and</br>
use an appropriate title for the figure. Consider displaying either</br>
the x-axis, the y-axis, or both on the log-scale. You can do this</br>
by using the ax.set_xscale() and ax.set_yscale() methods of .set_xscale() .set_yscale()</br>
the axes you are plotting to.</br>

```python
import matplotlib.pyplot as plt

# Generate x and y values
x_values = range(1, 11)
y_values = [Power3(x, 2) for x in x_values]

# Create the plot
fig, ax = plt.subplots()
ax.plot(x_values, y_values, marker='o', label=f'x^2')

# Customize the plot
ax.set_xlabel("x")
ax.set_ylabel("f(x) = x^2")
ax.set_title("Plot of f(x) = x^2")

# Set x-axis to log scale
ax.set_xscale('log')
ax.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

def PlotPower(x_values, a):
    y_values = [Power3(x, a) for x in x_values]
    fig, ax = plt.subplots()
    ax.plot(x_values, y_values, marker='o', label=f'x^{a}')
    ax.set_xlabel("x")
    ax.set_ylabel(f"f(x) = x^{a}")
    ax.set_title(f"Plot of f(x) = x^{a}")
    ax.legend()
    plt.show()

PlotPower(np.arange(1, 11), 3)
     
```

(d) Produce some of diagnostic plots of the linear regression fit as</br>
described in the lab. Comment on any problems you see with the</br>
fit. Do the residual plots suggest any unusually large outliers?</br>
Does the leverage plot identify any observations with unusually</br>
high leverage?</br>

```python
import statsmodels.api as sm
import matplotlib.pyplot as plt
import numpy as np

X = sm.add_constant(Auto[['cylinders','displacement','horsepower',
                          'weight','acceleration','year','origin']])
y = Auto['mpg']

model = sm.OLS(y, X).fit()

# Diagnostic Plots
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Diagnostic Plots for Multiple Linear Regression Model', fontsize=14, fontweight='bold')

# --- (1) Residuals vs Fitted ---
axes[0, 0].scatter(model.fittedvalues, model.resid, alpha=0.6)
axes[0, 0].axhline(y=0, color='red', linestyle='--')
axes[0, 0].set_xlabel('Fitted values')
axes[0, 0].set_ylabel('Residuals')
axes[0, 0].set_title('Residuals vs Fitted')
axes[0, 0].grid(True)

# --- (2) Normal Q-Q ---
sm.qqplot(model.resid, line='45', fit=True, ax=axes[0, 1])
axes[0, 1].set_title('Normal Q-Q')

# --- (3) Scale-Location Plot ---
# y軸 = √|standardized residuals|
standardized_resid = model.get_influence().resid_studentized_internal
axes[1, 0].scatter(model.fittedvalues, np.sqrt(np.abs(standardized_resid)), alpha=0.6)
axes[1, 0].set_xlabel('Fitted values')
axes[1, 0].set_ylabel('√|Standardized residuals|')
axes[1, 0].set_title('Scale-Location')
axes[1, 0].grid(True)

# --- (4) Residuals vs Leverage ---
influence = model.get_influence()
leverage = influence.hat_matrix_diag
axes[1, 1].scatter(leverage, standardized_resid, alpha=0.6)
axes[1, 1].axhline(y=0, color='red', linestyle='--')
axes[1, 1].set_xlabel('Leverage')
axes[1, 1].set_ylabel('Standardized residuals')
axes[1, 1].set_title('Residuals vs Leverage')
axes[1, 1].grid(True)

# Cook's distance 參考線
cooks_d, _ = influence.cooks_distance
axes[1, 1].scatter(leverage, standardized_resid, s=1000*cooks_d, alpha=0.2, color='orange', label="Cook's D (size)")
axes[1, 1].legend()

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

(1) Residuals vs Fitted（左上）：檢查是否存在非線性關係或變異數不均。看「預測錯多少」會不會有規律。
預測誤差（Residual）平均在 0 附近。</br>
殘差大致圍繞在紅線（0）附近，但中間區域略呈「U 型」，代表模型可能沒抓到某種非線性關係。</br>
(2) Normal Q–Q（右上）：檢查殘差是否服從常態分布。</br>
大多數點沿著紅線排列，說明常態性大致成立；</br>
右上角少數點偏離紅線，代表存在少量高正殘差的 outlier；</br>
沒有系統性的偏移（例如整體 S 形）。</br>
(3) Scale–Location（左下）：看「預測誤差的大小」是不是平均，檢查殘差的變異是否隨 fitted value 改變（變異數 heteroscedasticity）。</br>
散點分布大致均勻；右側的殘差稍微分散較大，表示在高 mpg 區誤差越大，高mpg 會預測不準</br>
(4) Residuals vs Leverage（右下）：檢查是否存在高影響值觀測點（influential points）。</br>
橘色點的大小代表 Cook’s Distance（觀測值影響整體模型的程度）。</br>
大多數點集中在低槓桿區（左側，Leverage < 0.05）；</br>
少數右下角的點（Leverage ≈ 0.17）有明顯高 Cook’s D；</br>
代表有少數樣本（可能是極端車型）對模型影響顯著。</br>


(e) Fit some models with interactions as described in the lab. Do
any interactions appear to be statistically significant?
兩個變數交互關係（horsepower＊weight）對 mpg 。

```python
import statsmodels.formula.api as smf
import pandas as pd

# 交互組合
test = [
    ('weight', 'year'),
    ('cylinders', 'displacement'),
    ('displacement', 'weight'),
    ('acceleration', 'weight')
]

interaction_results = []

for var1, var2 in test:
    formula = f'mpg ~ cylinders + displacement + horsepower + weight + acceleration + year + origin + {var1}*{var2}'
    model = smf.ols(formula=formula, data=Auto).fit()
    term_name = f'{var1}:{var2}'  
    # 取出交互作用係數與 p-value
    if term_name in model.pvalues:
        p_value = model.pvalues[term_name]
        coef = model.params[term_name]
        interaction_results.append({
            'Interaction': term_name,
            'Coefficient': coef,
            'p-value': p_value,
            'Significant (p<0.05)': 'Yes' if p_value < 0.05 else 'No'
        })

interaction_df = pd.DataFrame(interaction_results).sort_values('p-value')
print(interaction_df.to_string(index=False))
```

p < 0.05 顯示變數間存在聯合影響。</br>
「排氣量 × 重量」及「汽缸數 × 排氣量」顯示雙重放大油耗惡化的效果，</br>
「重量 × 年份」則反映新車技術改善使重量影響減弱。</br>

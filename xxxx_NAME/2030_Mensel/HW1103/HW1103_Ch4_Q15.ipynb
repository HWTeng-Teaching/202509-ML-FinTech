QUESTION D

Now create a new function, Power3(), that actually returns the result x**a as a Python object, rather than simply printing it to the screen. That is, if you store the value x**a in an object called result within your function, then you can simply return this result, using the following line:

return result

Note that the line above should be the last line in your function, and it should be indented 4 spaces.

```python
def Power3(x, a):
    result = x**a
    return result
```

```python
x_vals = np.arange(1, 11)
y_vals = [Power3(x, 2) for x in x_vals]

plt.plot(x_vals, y_vals, marker='o')
plt.title('Plot of f(x) = x²')
plt.xlabel('x')
plt.ylabel('x²')
plt.grid(True)
plt.show()
```

```python
# Optional: log scale
plt.plot(x_vals, y_vals, marker = 'o')
plt.xscale('log')      # Log scale for x-axis
plt.yscale('log')      # Log scale for y-axis
plt.xlabel('x (log scale)')
plt.ylabel('x² (log scale)')
plt.title('Log-Log Plot of x²')
plt.grid(True)
plt.show()
```

Create a function, PlotPower(), that allows you to create a plot of x against x**a for a fixed a and a sequence of values of x. For instance, if you call

PlotPower(np.arange(1, 11), 3)

then a plot should be created with an x-axis taking on values 1, 2, . . . , 10, and a y-axis taking on values 13, 23, . . . , 103.

```python
def PlotPower(x_values, a):
    y_values = [Power3(x, a) for x in x_values]
    
    plt.plot(x_values, y_values, marker = 'o')
    plt.title(f'Plot of f(x) = x^{a}')
    plt.xlabel('x')
    plt.ylabel(f'x^{a}')
    plt.grid(True)
    plt.show()

# Example usage:
PlotPower(np.arange(1, 11), 3)
```

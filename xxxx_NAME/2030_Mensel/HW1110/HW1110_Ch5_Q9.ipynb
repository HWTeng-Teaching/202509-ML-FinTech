```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from ISLP import load_data

boston = load_data("Boston")
```

QUESTION A

Based on this data set, provide an estimate for the population mean of medv. Call this estimate µˆ.

```python
# Assume df is loaded and has column 'medv'
medv = boston['medv'].to_numpy()
n = len(medv)

mu_hat = np.mean(medv)

print(f"Mean (mu_hat): {mu_hat:.3f}")
```

QUESTION B

Provide an estimate of the standard error of µˆ. Interpret this result. 
Hint: We can compute the standard error of the sample mean by dividing the sample standard deviation by the square root of the number of observations.

```python
se_mean_analytic = np.std(medv, ddof=1) / np.sqrt(n)
print(f"SE(mean) analytic: {se_mean_analytic:.3f}")
```

Interpretation:

This value quantifies the uncertainty in our estimate of the population mean 𝜇^. It tells us that if we repeatedly sampled 100 observations from the same population and computed the mean each time, the sample means would typically vary by about 0.409 units around the true population mean.

QUESTION C

Now estimate the standard error of µˆ using the bootstrap. How does this compare to your answer from (b)?

```python
rng = np.random.default_rng(0)
B = 1000
boot_means = np.empty(B)

for b in range(B):
    sample = rng.choice(medv, size=n, replace=True)
    boot_means[b] = np.mean(sample)

se_mean_boot = np.std(boot_means, ddof=1)

print(f"SE(mean) bootstrap: {se_mean_boot:.3f}")
```

The standard error estimate of µˆ using bootstrap give a slightly higher value than the non-bootstrapped one.

QUESTION D

Based on your bootstrap estimate from (c), provide a 95 % confidence interval for the mean of medv. Compare it to the results obtained by using Boston['medv'].std() and the two standard error rule (3.9).
Hint: You can approximate a 95 % confidence interval using the formula [ˆ µ − 2SE(ˆ µ), µˆ + 2SE(ˆ µ)].

```python
ci_two_se = (mu_hat - 2*se_mean_boot, mu_hat + 2*se_mean_boot)
ci_percentile = (np.percentile(boot_means, 2.5), np.percentile(boot_means, 97.5))


print(f"95% CI (two-SE rule): [{ci_two_se[0]:.3f}, {ci_two_se[1]:.3f}]")
print(f"95% CI (bootstrap percentile): [{ci_percentile[0]:.3f}, {ci_percentile[1]:.3f}]")
```

They are almost overlapping and narrow, indicating a precise estimate of the population mean. 

QUESTION E

Based on this data set, provide an estimate, µˆmed, for the median value of medv in the population.

```python
mu_med_hat = np.median(medv)

print(f"Median (mu_med_hat): {mu_med_hat:.3f}")
```

QUESTION F

We now would like to estimate the standard error of µˆmed. Unfortunately, there is no simple formula for computing the standard error of the median. Instead, estimate the standard error of the median using the bootstrap. Comment on your findings.

```python
boot_medians = np.empty(B)
for b in range(B):
    sample = rng.choice(medv, size=n, replace=True)
    boot_medians[b] = np.median(sample)

se_median_boot = np.std(boot_medians, ddof=1)
ci_median_percentile = (np.percentile(boot_medians, 2.5), np.percentile(boot_medians, 97.5))

print(f"SE(median) bootstrap: {se_median_boot:.3f}")
print(f"95% CI median (bootstrap percentile): [{ci_median_percentile[0]:.3f}, {ci_median_percentile[1]:.3f}]")
```

Mean (22.533) and median (21.200) are close, indicating a normality and robustness to outliers. The confidence interval of median and mean overlap just a little.

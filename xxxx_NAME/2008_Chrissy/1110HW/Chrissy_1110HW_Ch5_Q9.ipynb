We will now consider the Boston housing data set, from the ISLP library.

### (a)  
Based on this data set, provide an estimate for the population mean of \( \text{medv} \). Call this estimate \( \hat{\mu} \).

```python
from ISLP import load_data
import numpy as np

Boston = load_data('Boston')
Boston
```

```python
mu_hat = np.mean(Boston['medv'])
mu_hat
```

### (b)  
Provide an estimate of the standard error of \( \hat{\mu} \). Interpret this result.  
Hint: We can compute the standard error of the sample mean by dividing the sample standard deviation by the square root of the number of observations.

```python
se_mu_hat = np.std(Boston['medv'], ddof=1) / np.sqrt(len(Boston['medv']))
se_mu_hat
```

### (c)  
Now estimate the standard error of \( \hat{\mu} \) using the bootstrap. How does this compare to your answer from (b)?

```python
rng = np.random.default_rng(1)
boot_means = [np.mean(rng.choice(Boston['medv'], size=len(Boston['medv']), replace=True)) for _ in range(1000)]
boot_se = np.std(boot_means, ddof=1)
boot_se
```

**ANS for question c:**  
The standard error of hat mu from bootstrap (0.4088) is quite the same as we have in question b (0.4155). 

### (d)  
Based on your bootstrap estimate from (c), provide a 95% confidence interval for the mean of \( \text{medv} \). Compare it to the results obtained by using `Boston['medv'].std()` and the two standard error rule (3.9).  
Hint: You can approximate a 95% confidence interval using the formula \([ \hat{\mu} - 2SE(\hat{\mu}), \hat{\mu} + 2SE(\hat{\mu}) ]\).

```python
ci_lower = mu_hat - 2 * boot_se
ci_upper = mu_hat + 2 * boot_se
(ci_lower, ci_upper)
```

The 95% confidence interval for this mean of `medv` is [21.7, 23.3]

### (e)

We now would like to estimate the standard error of \( \hat{\mu}_{\text{med}} \). Unfortunately, there is no simple formula for computing the standard error of the median. Instead, estimate the standard error of the median using the bootstrap. Comment on your findings.

```python
import numpy as np
import pandas as pd
from ISLP import load_data

# 讀資料
Boston = load_data('Boston')
medv = Boston['medv'].values
n = len(medv)

print("Sample size n =", n)
```

```python
mu_hat = medv.mean()
print("\n(a) Sample mean (mu_hat) =", mu_hat)
```

```python
SE_formula = medv.std(ddof=1) / np.sqrt(n)
print("\n(b) Standard error (formula) =", SE_formula)
```

```python
B = 10000
rng = np.random.default_rng(1)
boot_means = np.zeros(B)

for b in range(B):
    sample = rng.choice(medv, size=n, replace=True)
    boot_means[b] = sample.mean()

SE_boot_mean = boot_means.std(ddof=1)
print("\n(c) Bootstrap SE (mean) =", SE_boot_mean)
```

```python
CI_lower = mu_hat - 2 * SE_boot_mean
CI_upper = mu_hat + 2 * SE_boot_mean
print("\n(d) Bootstrap 95% CI for mean = [", CI_lower, ",", CI_upper, "]")

# 兩倍標準誤 rule of thumb
CI_formula_lower = mu_hat - 2 * SE_formula
CI_formula_upper = mu_hat + 2 * SE_formula
print("   Formula-based CI =", [CI_formula_lower, CI_formula_upper])
```

```python
mu_med = np.median(medv)
print("\n(e) Sample median =", mu_med)
```

```python
boot_medians = np.zeros(B)

for b in range(B):
    sample = rng.choice(medv, size=n, replace=True)
    boot_medians[b] = np.median(sample)

SE_boot_median = boot_medians.std(ddof=1)
print("\n(f) Bootstrap SE (median) =", SE_boot_median)
```

```python
mu_01 = np.percentile(medv, 10)
print("\n(g) 10th percentile =", mu_01)
```

```python
boot_p10 = np.zeros(B)

for b in range(B):
    sample = rng.choice(medv, size=n, replace=True)
    boot_p10[b] = np.percentile(sample, 10)

SE_boot_p10 = boot_p10.std(ddof=1)
print("\n(h) Bootstrap SE (10th percentile) =", SE_boot_p10)
```

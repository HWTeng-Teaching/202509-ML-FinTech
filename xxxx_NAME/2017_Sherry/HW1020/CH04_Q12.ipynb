### Suppose that you wish to classify an observation $X\in\mathbb R$ into apple and orange. You fit a logistic regression model and find that
$$\widehat{\text{Pr}}(Y=\texttt{orange}|X = x) = \frac{\exp(\hat\beta_0 + \hat\beta_1x)}{1+\exp(\hat\beta_0 + \hat\beta_1x)}$$
Your friend input the above logistic regression model to the same data using the softmax formulation in (4.13):
$$\widehat{\text{Pr}}(Y=\texttt{orange}|X = x) = \frac{\exp(\hat\alpha_{\texttt{orange}0} + \hat\alpha_{\texttt{orange}1}x)}{\exp(\hat\alpha_{\texttt{orange}0} + \hat\alpha_{\texttt{orange}1}x) + \exp(\hat\alpha_{\texttt{apple}0} + \hat\alpha_{\texttt{apple}1}x)}$$

###(a) What is the log odds of orange versus apple in your model?

$$\text{odds} = \frac{\widehat{\text{Pr}}(Y=\texttt{orange}|X = x)}{1 - \widehat{\text{Pr}}(Y=\texttt{orange}|X = x)} = \exp(\hat\beta_0 + \hat\beta_1x),$$
log odds:
$$\log \exp(\hat\beta_0 + \hat\beta_1x) = \hat\beta_0 + \hat\beta_1x$$

---

###(b) What is the log odds of orange versus apple in your friend’s model?

$$\text{odds} = \frac{\widehat{\text{Pr}}(Y=\texttt{orange}|X = x)}{1 - \widehat{\text{Pr}}(Y=\texttt{orange}|X = x)}  = \exp(\hat\alpha_{\texttt{orange}0} - \hat\alpha_{\texttt{apple}0} + (\hat\alpha_{\texttt{orange}1} - \hat\alpha_{\texttt{apple}1})x),$$
log odds:
$$\log\exp(\hat\alpha_{\texttt{orange}0} - \hat\alpha_{\texttt{apple}0} + (\hat\alpha_{\texttt{orange}1} - \hat\alpha_{\texttt{apple}1})x) = \hat\alpha_{\texttt{orange}0} - \hat\alpha_{\texttt{apple}0} + (\hat\alpha_{\texttt{orange}1} - \hat\alpha_{\texttt{apple}1})x$$

---

(c) Suppose that in your model, $\hat\beta_0 = 2$ and $\hat\beta_1 = -1$. What are the coefficient estimates in your friend’s model? Be as specific as possible.


In fact, when catrgory count equals to 2, the *softmax* activation function is some sort of the *sigmoid* activation function with different parameters.
$$\widehat{\text{Pr}}(Y=\texttt{orange}|X = x) = \frac{\exp(\hat\alpha_{\texttt{orange}0} + \hat\alpha_{\texttt{orange}1}x)}{\exp(\hat\alpha_{\texttt{orange}0} + \hat\alpha_{\texttt{orange}1}x) + \exp(\hat\alpha_{\texttt{apple}0} + \hat\alpha_{\texttt{apple}1}x)} = \frac{1}{1+\exp[(\hat\alpha_{\texttt{apple}0} - \hat\alpha_{\texttt{orange}0})+(\hat\alpha_{\texttt{apple}1}-\hat\alpha_{\texttt{orange}1})x]}$$

By comparing the parameters, we get:
$$-\hat\beta_0 = -2 =  \hat\alpha_{\texttt{apple}0} - \hat\alpha_{\texttt{orange}0}$$
and
$$-\hat\beta_1 = 1 = \hat\alpha_{\texttt{apple}1}-\hat\alpha_{\texttt{orange}1}$$

This time, your friend gets the coefficient estimates $\hat\alpha_{\texttt{apple}0} = 1.2, \hat\alpha_{\texttt{apple}1} = −2, \hat\alpha_{\texttt{orange}0} = 3, \hat\alpha_{\texttt{orange}1} = 0.6$. What are the coefficient estimates in your model?


From the derivation we get from (c), we must have:

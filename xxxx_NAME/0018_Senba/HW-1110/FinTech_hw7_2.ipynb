# 9. We will now consider the Boston housing data set, from the ISLP library.

```python
import numpy as np
import pandas as pd
from ISLP import load_data
from sklearn.utils import resample

Boston = load_data('Boston')
Boston["medv"].describe()
```

## (a) Based on this data set, provide an estimate for the population mean of medv. Call this estimate µˆ.

```python
mu_hat = Boston['medv'].mean()
print(f"{mu_hat:.4f}")
```

## (b) Provide an estimate of the standard error of µˆ. Interpret this result.
$$\text{SE}(\hat{\mu}) = \frac{\sigma}{\sqrt{n}} \approx \frac{s}{\sqrt{n}}$$

$\text{我們抽樣的結果應該落在 }μ̂ ± SE\text{ 的範圍內}。$

```python
sample_std = Boston['medv'].std()
n = len(Boston)
se_formula = sample_std / np.sqrt(n)

print(f"{se_formula:.4f}")
```

## (c) Now estimate the standard error of µˆ using the bootstrap. Howdoes this compare to your answer from (b)?

### Result
0.4089 與 0.4183 相對很接近，代表當樣本數 $n$ 夠大，可能可以直接選擇使用$\text{ }SE$。

```python
np.random.seed(34)
B = 1000
bootstrap_means = []

for _ in range(B):
    bootstrap_sample = Boston['medv'].sample(n=n, replace=True)
    bootstrap_means.append(bootstrap_sample.mean())

se_bootstrap = np.std(bootstrap_means, ddof=1)
print(f"{se_bootstrap:.4f}")
```

## (d) Based on your bootstrap estimate from (c), provide a 95 % confdence interval for the mean of medv. Compare it to the results obtained by using Boston['medv'].std() and the two standard error rule.

### Result
與 (c) 得到的結論相近，兩者的信賴區間（Confidence Interval）非常相近。

```python
lower_bound_bootstrap = mu_hat - 2 * se_bootstrap
upper_bound_bootstrap = mu_hat + 2 * se_bootstrap
lower_bound_formula = mu_hat - 2 * se_formula
upper_bound_formula = mu_hat + 2 * se_formula

print(f"Confidence Interval (bootstrap): [{lower_bound_bootstrap:.4f}, {upper_bound_bootstrap:.4f}]")
print(f"Confidence Interval (std())): [{lower_bound_formula:.4f}, {upper_bound_formula:.4f}]")
```

## (e) Based on this data set, provide an estimate, $\hat{\mu}_{med}$, for the medianvalue of medv in the population.

```python
mu_hat_med = Boston['medv'].median()
print(f"{mu_hat_med:.4f}")
```

## (f) We now would like to estimate the standard error of $\hat{\mu}_{med}$. Unfortunately, there is no simple formula for computing the standard error of the median. Instead, estimate the standard error of the median using the bootstrap. Comment on your fndings.

### Comment
中位數的 SE（0.3881） < 平均數的 SE（0.4183），這可能是因為中位數取樣範圍較小，波動性較高


```python
np.random.seed(34)
bootstrap_medians = []

for _ in range(B):
    bootstrap_sample = Boston['medv'].sample(n=n, replace=True)
    bootstrap_medians.append(bootstrap_sample.median())

se_median_bootstrap = np.std(bootstrap_medians, ddof=1)

print(f"{se_median_bootstrap:.4f}")
```

## (g) Based on this data set, provide an estimate for the tenth per￾centile of medv in Boston census tracts. Call this quantity $\hat{\mu}_{0.1}$.

```python
mu_hat_0_1 = np.percentile(Boston['medv'], 10)
print(f"{mu_hat_0_1:.4f}")
```

## (h) Use the bootstrap to estimate the standard error of µˆ0.1. Comment on your fndings.

### Comment

第十百分位數的 SE（0.5066） 遠比平均數和中位數的 SE 大，因為它們更多地取決於分佈的尾部，這部分在小樣本中更不穩定。

```python
np.random.seed(34)
bootstrap_percentiles = []

for _ in range(B):
    bootstrap_sample = Boston['medv'].sample(n=n, replace=True)
    bootstrap_percentiles.append(np.percentile(bootstrap_sample, 10))

se_percentile_bootstrap = np.std(bootstrap_percentiles, ddof=1)

print(f"{se_percentile_bootstrap:.4f}")
```

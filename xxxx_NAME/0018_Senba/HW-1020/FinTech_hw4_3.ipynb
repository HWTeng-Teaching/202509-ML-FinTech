# Your friend fts a logistic regression model to the same data using the softmax formulation in (4.13), and fnds that

$$\text{Pr}(\hat{Y} = \text{orange}|X = x) = \frac{\exp(\hat{\alpha}_{\text{orange}, 0} + \hat{\alpha}_{\text{orange}, 1} x)}{\exp(\hat{\alpha}_{\text{orange}, 0} + \hat{\alpha}_{\text{orange}, 1} x) + \exp(\hat{\alpha}_{\text{apple}, 0} + \hat{\alpha}_{\text{apple}, 1} x)}$$

## (a) What is the log odds of orange versus apple in your model?

$$ \text{Odds}=\frac{\text{Probability of orange}}{\text{Probability of apple}} = \frac{\text{Pr}(\hat{Y} = \text{orange}|X = x)}{\text{Pr}(\hat{Y} = \text{apple}|X = x)} = \frac{\hat{P}}{1 - \hat{P}}$$

$$
\text{So We Get:}\quad \hat{P} = \frac{\exp(\hat{\beta}_0 + \hat{\beta}_1 x)}{1 + \exp(\hat{\beta}_0 + \hat{\beta}_1 x)} \quad \text{and} \quad 1 - \hat{P} = \frac{1}{1 + \exp(\hat{\beta}_0 + \hat{\beta}_1 x)}
$$

---

$$\text{Odds} = \exp(\hat{\beta}_0 + \hat{\beta}_1 x)$$

$$\text{Log Odds} = \ln(\text{Odds}) = \ln\left(\exp(\hat{\beta}_0 + \hat{\beta}_1 x)\right) = \mathbf{\hat{\beta}_0 + \hat{\beta}_1} x$$


## (b)  What is the log odds of orange versus apple in your friend’s model?

$$\text{Pr}(\hat{Y} = \text{orange}|X = x) = \frac{\exp(\hat{\alpha}_{\text{orange}, 0} + \hat{\alpha}_{\text{orange}, 1} x)}{\exp(\hat{\alpha}_{\text{orange}, 0} + \hat{\alpha}_{\text{orange}, 1} x) + \exp(\hat{\alpha}_{\text{apple}, 0} + \hat{\alpha}_{\text{apple}, 1} x)} $$
$$\text{Pr}(\hat{Y} = \text{apple}|X = x) = \frac{\exp(\hat{\alpha}_{\text{apple}, 0} + \hat{\alpha}_{\text{apple}, 1} x)}{\exp(\hat{\alpha}_{\text{orange}, 0} + \hat{\alpha}_{\text{orange}, 1} x) + \exp(\hat{\alpha}_{\text{apple}, 0} + \hat{\alpha}_{\text{apple}, 1} x)}$$

---

$$\text{Odds} = \frac{\text{Pr}(\hat{Y} = \text{orange}|X = x)}{\text{Pr}(\hat{Y} = \text{apple}|X = x)} = \text{Odds} = \frac{\exp(\hat{\alpha}_{\text{orange}, 0} + \hat{\alpha}_{\text{orange}, 1} x)}{\exp(\hat{\alpha}_{\text{apple}, 0} + \hat{\alpha}_{\text{apple}, 1} x)}$$

$$\text{Log Odds} = \ln\left(\frac{\exp(\hat{\alpha}_{\text{orange}, 0} + \hat{\alpha}_{\text{orange}, 1} x)}{\exp(\hat{\alpha}_{\text{apple}, 0} + \hat{\alpha}_{\text{apple}, 1} x)}\right) = (\hat{\alpha}_{\text{orange}, 0} + \hat{\alpha}_{\text{orange}, 1} x) - (\hat{\alpha}_{\text{apple}, 0} + \hat{\alpha}_{\text{apple}, 1} x) = \mathbf{(\hat{\alpha}_{\text{orange}, 0} - \hat{\alpha}_{\text{apple}, 0}) + (\hat{\alpha}_{\text{orange}, 1} - \hat{\alpha}_{\text{apple}, 1})}x$$

## (c) Suppose that in your model, $\hat{\beta}_0 = 2$ and $\hat{\beta}_1 = -1$. What are the coefficient estimates in your friend’s model? Be as specific as possible.

$$\text{Log Odds} = \ln(\text{Odds}) = \mathbf{\hat{\beta}_0 + \hat{\beta}_1}x  = \mathbf{(\hat{\alpha}_{\text{orange}, 0} - \hat{\alpha}_{\text{apple}, 0}) + (\hat{\alpha}_{\text{orange}, 1} - \hat{\alpha}_{\text{apple}, 1})}x$$

$$\text{So that}\quad \hat{\beta}_0 = \hat{\alpha}_{\text{orange}, 0} - \hat{\alpha}_{\text{apple}, 0} \quad\text{and}\quad \hat{\beta}_1 = \hat{\alpha}_{\text{orange}, 1} - \hat{\alpha}_{\text{apple}, 1}$$

```python
beta_0_hat = 2
beta_1_hat = -1

alpha_apple_0_hat = 0
alpha_apple_1_hat = 0

alpha_orange_0_hat = beta_0_hat + alpha_apple_0_hat
alpha_orange_1_hat = beta_1_hat + alpha_apple_1_hat

print(f"模型係數 (beta_0, beta_1): ({beta_0_hat}, {beta_1_hat})")

print("\n假設 apple 為 0：\n")
print(f"Alpha_apple_0: {0}")
print(f"Alpha_apple_1: {0}")
print(f"Alpha_orange_0: {alpha_orange_0_hat}")
print(f"Alpha_orange_1: {alpha_orange_1_hat}")

alpha_orange_0_hat = 0
alpha_orange_1_hat = 0

alpha_apple_0_hat = alpha_orange_0_hat - beta_0_hat
alpha_apple_1_hat = alpha_orange_1_hat - beta_1_hat

print("\n假設 orange 為 0：\n")
print(f"Alpha_apple_0: {alpha_apple_0_hat}")
print(f"Alpha_apple_1: {alpha_apple_1_hat}")
print(f"Alpha_orange_0: {0}")
print(f"Alpha_orange_1: {0}")
```

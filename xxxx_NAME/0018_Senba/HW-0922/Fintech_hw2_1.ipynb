```python
import numpy as np
from scipy.cluster.hierarchy import linkage, dendrogram, leaves_list
from scipy.spatial.distance import squareform
import matplotlib.pyplot as plt

D = np.array([
    [0.0, 0.30, 0.40, 0.70],
    [0.30, 0.0, 0.50, 0.80],
    [0.40, 0.50, 0.0, 0.45],
    [0.70, 0.80, 0.45, 0.0]
])
```

## On the basis of this dissimilarity matrix, sketch the dendrogram that results from hierarchically clustering these four observa￾tions using complete linkage. Be sure to indicate on the plot the height at which each fusion occurs, as well as the observations corresponding to each leaf in the dendrogram.

```python
condensed = squareform(D)
Z_complete = linkage(condensed, method='complete')

plt.figure(figsize=(6,4))
dendrogram(Z_complete, labels=['1','2','3','4'], orientation='top')
plt.title('Complete-linkage hierarchical clustering')
plt.ylabel('Dissimilarity (height)')
plt.axhline(0.30, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0.45, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0.80, color='gray', linestyle='--', linewidth=0.7)
plt.show()

print("\nLinkage matrix Z (format: idx1, idx2, height, sample_count):")
print(Z_complete)
```

## Repeat (a), this time using single linkage clustering.

```python
Z_single = linkage(condensed, method='single')

plt.figure(figsize=(6,4))
dendrogram(Z_single, labels=['1','2','3','4'], orientation='top')
plt.title('Single-linkage hierarchical clustering')
plt.ylabel('Dissimilarity (height)')
plt.axhline(0.30, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0.40, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0.45, color='gray', linestyle='--', linewidth=0.7)
plt.show()

print("\nLinkage matrix (single linkage):")
print(Z_single)
```

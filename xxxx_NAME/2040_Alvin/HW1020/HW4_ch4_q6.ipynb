 Suppose we collect data for a group of students in a statistics class
with variables $X_1$ = hours studied, $X_2$ = undergrad GPA, and $Y$ =
receive an A. We fit a logistic regression and produce estimated
coefficient, 
$\hat{\beta}_0 = -6;\;\hat{\beta}_1 = 0.05;\;\hat{\beta}_2 = 1$.


(a) Estimate the probability that a student who studies for $40$ h and has an undergrad GPA of $3.5$ gets an A in the class.


In the given context, we know our model is as the form:
$$P(\hat Y = 1| \mathbf{X}) = \frac{1}{1+ \exp(-0.05x_1 - x_2 +6)}$$
Which means if we take $x_1=40$ and $x_2 = 3.5$, we get:
$$P(\hat Y = 1| (x_1,\;x_2) = (40,\;3.5)) = \frac{1}{1+\exp(-0.05*40-3.5+6)}\approx \boxed{0.3775}$$

---

(b) How many hours would the student in part (a) need to study to
have a $50 \%$ chance of getting an A in the class?

We now need to solve for $x_1^{\ast}$ such that the following equation holds:
$$P(\hat Y|(x_1,\;x_2) = (40,\;x_1^{\ast}))= 0.5$$
We can easily oberseve that:
$$P(\hat Y|(x_1,\;x_2)) = 0.5\iff \exp(-0.05x_1-x_2+6) = 1 \iff 0.05x_1+x_2 - 6 = 0$$
Hence we have the equation:
$$0.05x_1^{\ast}+3.5 -6=0,$$
whcih yields the result:
$$\boxed{x_1^{\ast} = 50}$$

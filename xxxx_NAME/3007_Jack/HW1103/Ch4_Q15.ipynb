(d)

```python
def Power3(x, a):
    result = x ** a
    return result
```

```python
import matplotlib.pyplot as plt

x = list(range(1, 11))
y = [Power3(i, 2) for i in x]  

fig, ax = plt.subplots()
ax.plot(x, y, marker='o', color='blue')
ax.set_title("f(x) = x^2")
ax.set_xlabel("x")
ax.set_ylabel("x squared")

plt.grid(True)
plt.show()
```

```python
import numpy as np
def PlotPower(x, a):
    y = x ** a
    plt.plot(x, y, marker='o', color='purple')
    plt.title(f"f(x) = x^{a}")
    plt.xlabel("x")
    plt.ylabel(f"x^{a}")
    plt.grid(True)
    plt.show()

PlotPower(np.arange(1, 11), 3)
```

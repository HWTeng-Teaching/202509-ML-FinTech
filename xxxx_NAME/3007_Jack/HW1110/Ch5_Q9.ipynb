```python
import numpy as np
import pandas as pd
Boston = pd.read_csv("Boston.csv")
Boston
```

```python
Boston['medv'].mean()
```

(b)

```python
medv = Boston['medv']
n = len(medv)

se_mu = medv.std(ddof=1) / np.sqrt(n)
se_mu
```

(c)

```python
B = 1000
boot_means = []

rng = np.random.default_rng(1)

for _ in range(B):
    sample = rng.choice(medv, size=n, replace=True)
    boot_means.append(sample.mean())

boot_se = np.std(boot_means, ddof=1)
boot_se
```

(d)

```python
mu_hat = medv.mean()
ci_lower = mu_hat - 2 * boot_se
ci_upper = mu_hat + 2 * boot_se
(ci_lower, ci_upper)
```

(e)

```python
mu_med = np.median(medv)
mu_med
```

(f)

```python
B = 1000
rng = np.random.default_rng(1)
boot_medians = []

for _ in range(B):
    sample = rng.choice(medv, size=n, replace=True)
    boot_medians.append(np.median(sample))

boot_medians = np.array(boot_medians)
se_med_boot = boot_medians.std(ddof=1)
print("Bootstrap SE(μ̂_med):", se_med_boot)
```

(g)

```python
mu_0_1 = np.percentile(medv, 10)   
mu_0_1
```

(h)

```python
B = 1000
rng = np.random.default_rng(2)
boot_p10 = []

for _ in range(B):
    sample = rng.choice(medv, size=n, replace=True)
    boot_p10.append(np.percentile(sample, 10))

boot_p10 = np.array(boot_p10)
se_p10_boot = boot_p10.std(ddof=1)
print("Bootstrap SE(μ̂_0.1):", se_p10_boot)
```

# Suppose that we have four observations, for which we compute a dissimilarity matrix, given by

# (matrix).

# For instance, the dissimilarity between the first and second observations is 0.3, and the dissimilarity between the second and fourth observations is 0.8.

```python
import numpy as np
from scipy.cluster.hierarchy import linkage, dendrogram, leaves_list
from scipy.spatial.distance import squareform
import matplotlib.pyplot as plt

D = np.array([
    [0.0, 0.30, 0.40, 0.70],
    [0.30, 0.0, 0.50, 0.80],
    [0.40, 0.50, 0.0, 0.45],
    [0.70, 0.80, 0.45, 0.0]
])
```

# (a) On the basis of this dissimilarity matrix, sketch the dendrogram that results from hierarchically clustering these four observations using complete linkage. Be sure to indicate on the plot the height at which each fusion occurs, as well as the observations corresponding to each leaf in the dendrogram.

```python
# (a)
condensed = squareform(D, checks = True)
Z_complete = linkage(condensed, method = 'complete') # complete linkage: max(a, b)

plt.figure(figsize=(6,4))
dendrogram(Z_complete, labels=['1', '2', '3', '4'], orientation='top')
plt.title('complete linkage hierarchical clustering')
plt.ylabel('dissimilarity')
plt.axhline(0.30, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0.45, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0.80, color='gray', linestyle='--', linewidth=0.7)
plt.show()

print("\nLinkage matrix Z (format: idx1, idx2, height, sample_count):")
print(Z_complete)
```

# (b) Repeat (a), this time using single linkage clustering.

```python
# (b)
Z_complete = linkage(condensed, method = 'single') # single linkage: min(a, b)

plt.figure(figsize=(6,4))
dendrogram(Z_complete, labels=['1', '2', '3', '4'], orientation='top')
plt.title('single linkage hierarchical clustering')
plt.ylabel('dissimilarity')
plt.axhline(0.30, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0.40, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0.45, color='gray', linestyle='--', linewidth=0.7)
plt.show()

print("\nLinkage matrix Z (format: idx1, idx2, height, sample_count):")
print(Z_complete)
```

#(c) Suppose that we cut the dendrogram obtained in (a) such that two clusters result. Which observations are in each cluster?

(c)\
When we cut the complete-linkage dendrogram from part (a) at a height between 0.45 and 0.80 (for example, at 0.46), we obtain two clusters:\
	•	Cluster 1: {1, 2}\
	•	Cluster 2: {3, 4}


# (d) Suppose that we cut the dendrogram obtained in (b) such that two clusters result. Which observations are in each cluster?

(d)\
Using the single-linkage dendrogram from part (b), cut at any height between 0.40 and 0.45 (e.g., 0.41). This yields two clusters:\
	•	Cluster 1: {1, 2, 3}\
	•	Cluster 2: {4}

# (e) It is mentioned in this chapter that at each fusion in the dendrogram, the position of the two clusters being fused can be swapped without changing the meaning of the dendrogram. Draw a dendrogram that is equivalent to the dendrogram in (a), for which two or more of the leaves are repositioned, but for which the meaning of the dendrogram is the same.

```python
# (e)
condensed = squareform(D, checks = True)
Z_complete = linkage(condensed, method = 'complete')

plt.figure(figsize=(6,4))
dendrogram(Z_complete, labels=['2', '1', '4', '3'], orientation='top')
plt.title('complete linkage hierarchical clustering')
plt.ylabel('dissimilarity')
plt.axhline(0.30, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0.45, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0.80, color='gray', linestyle='--', linewidth=0.7)
plt.show()
```

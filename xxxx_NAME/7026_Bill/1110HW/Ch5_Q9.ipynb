9. We will now consider the Boston housing data set, from the ISLP library.

```python
import numpy as np
import pandas as pd
from ISLP import load_data
from sklearn.utils import resample

Boston = load_data('Boston')
Boston["medv"].describe()
```

(a) Based on this data set, provide an estimate for the population mean of medv. Call this estimate µˆ.

```python
mu_hat = Boston['medv'].mean()
print(f"{mu_hat:.4f}")
```

(b) Provide an estimate of the standard error of µˆ. Interpret this result.

```python
#(b)
sample_std = Boston['medv'].std()
n = len(Boston)
se_formula = sample_std / np.sqrt(n)

print(f"{se_formula:.4f}")
```

(c) Now estimate the standard error of µˆ using the bootstrap. Howdoes this compare to your answer from (b)?

```python
#(c)
np.random.seed(34)
B = 1000
bootstrap_means = []

for _ in range(B):
    bootstrap_sample = Boston['medv'].sample(n=n, replace=True)
    bootstrap_means.append(bootstrap_sample.mean())

se_bootstrap = np.std(bootstrap_means, ddof=1)
print(f"{se_bootstrap:.4f}")
```

(d)

Based on your bootstrap estimate from (c), provide a 95 % confdence interval for the mean of medv. Compare it to the results obtained by using Boston['medv'].std() and the two standard error rule.

```python
#(d)
lower_bound_bootstrap = mu_hat - 2 * se_bootstrap
upper_bound_bootstrap = mu_hat + 2 * se_bootstrap
lower_bound_formula = mu_hat - 2 * se_formula
upper_bound_formula = mu_hat + 2 * se_formula

print(f"Confidence Interval (bootstrap): [{lower_bound_bootstrap:.4f}, {upper_bound_bootstrap:.4f}]")
print(f"Confidence Interval (std())): [{lower_bound_formula:.4f}, {upper_bound_formula:.4f}]")
```

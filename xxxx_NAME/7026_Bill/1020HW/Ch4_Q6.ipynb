6. Suppose we collect data for a group of students in a statistics class with variables X1 = hours studied, X2 = undergrad GPA, and Y = receive an A. We fit a logistic regression and produce estimated coefficients, β̂₀ = -6, β̂₁ = 0.05, and β̂₂ = 1.

```python
import numpy as np
```

(a) Estimate the probability that a student who studies for 40 h and has an undergrad GPA of 3.5 gets an A in the class.

```python
#(a)
beta_0_hat = -6
beta_1_hat = 0.05
beta_2_hat = 1

X1 = 40
X2 = 3.5


P_hat = np.exp(beta_0_hat + beta_1_hat * X1 + beta_2_hat * X2) / (1 + np.exp(beta_0_hat + beta_1_hat * X1 + beta_2_hat * X2))
print(f"study time: 40 hours, GPA: 3.5, probability :{P_hat * 100:.2f}%")
```

(b) How many hours would the student in part (a) need to study to have a 50% chance of getting an A in the class?

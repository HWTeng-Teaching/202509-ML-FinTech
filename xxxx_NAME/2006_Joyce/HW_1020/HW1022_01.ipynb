Suppose we collect data for a group of students in a statistics class with variables X1 = hours studied, X2 = undergrad GPA, and Y = receive an A. We fit a logistic regression and produce estimated coefficient, β_0 = −6, β_1 = 0.05, β_2 = 1.

(a) Estimate the probability that a student who studies for 40 h and has an undergrad GPA of 3.5 gets an A in the class.

```python
import numpy as np

# Define the coefficients
beta_0 = -6
beta_1 = 0.05
beta_2 = 1

# Define the student's data
hours_studied = 40
undergrad_gpa = 3.5

# Calculate z
z = beta_0 + beta_1 * hours_studied + beta_2 * undergrad_gpa

# Calculate the probability using the sigmoid function
probability = 1 / (1 + np.exp(-z))

# Print the result
print(f"The estimated probability of getting an A is: {probability:.4f}")
```

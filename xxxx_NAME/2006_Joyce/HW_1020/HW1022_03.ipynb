***Suppose that you wish to classify an observation X ∈ R into apples and oranges. You fit a logistic regression model and find that***

$P(Y=\text{orange}|X=x)= \frac{\exp(\hat{\beta}_0+\hat{\beta}_1x)}{1 + \exp(\hat{\beta}_0 + \hat{\beta}_1x)}$

***Your friend fits a logistic regression model to the same data using the softmax formulation in (4.13), and finds that***

$P(Y = \text{orange}|X = x) = \frac{\exp(\hat{\alpha}_{\text{orange}0} + \hat{\alpha}_{\text{orange}1}x)}{\exp(\hat{\alpha}_{\text{orange}0} + \hat{\alpha}_{\text{orange}1}x) + \exp(\hat{\alpha}_{\text{apple}0} + \hat{\alpha}_{\text{apple}1}x)}$

***(a) What is the log odds of orange versus apple in your model?***

理解對數勝算

對數勝算是勝算的自然對數。 由於勝算是事件發生的機率與事件不發生的機率之比，因此對數勝算可以表示為：

$log\ odds = ln(\frac{P(事件)}{1 - P(事件)})$

應用於您的模型

在您的模型中，$P(Y=橘子 | X=x) = \frac{exp(\hat{β}_0 + \hat{β}_1x)}{1 + exp(\hat{β}_0 + \hat{β}_1x)}$

因此，$1 - P(Y=橘子 | X=x) = 1 - \frac{exp(\hat{β}_0 + \hat{β}_1x)}{1 + exp(\hat{β}_0 + \hat{β}_1x)}$
$= \frac{(1 + exp(\hat{β}_0 + \hat{β}_1x) - exp(\hat{β}_0 + \hat{β}_1x))}{1 + exp(\hat{β}_0 + \hat{β}_1x)}$
$= \frac{1}{1 + exp(\hat{β}_0 + \hat{β}_1x)}$  這代表在給定 X=x 的情況下，蘋果的機率。

因此，橘子對蘋果的勝算是
$odds = \frac{P(Y=橘子 | X=x)}{P(Y=蘋果 | X=x)}$
$= \frac{\frac{exp(\hat{β}_0 + \hat{β}_1x)}{1 + exp(\hat{β}_0 + \hat{β}_1x)}}{\frac{1}{1 + exp(\hat{β}_0 + \hat{β}_1x)}}$
$= exp(\hat{β}_0 + \hat{β}_1x)$

現在，我們取勝算的自然對數 (ln) 以獲得對數勝算：

$log\ odds = ln(exp(\hat{β}_0 + \hat{β}_1x)) = \hat{β}_0 + \hat{β}_1x$

答案

在您的模型中，橘子對蘋果的對數勝算是 $\hat{β}_0 + \hat{β}_1x$。

```python
import numpy as np

def log_odds(beta_0, beta_1, x):
  """
  Calculates the log odds of orange versus apple in a logistic regression model.

  Args:
    beta_0: The intercept coefficient.
    beta_1: The coefficient for the predictor variable x.
    x: The value of the predictor variable.

  Returns:
    The log odds of orange versus apple.
  """
  return beta_0 + beta_1 * x

# Example Usage:
beta_0 = -2.0  # Example intercept coefficient
beta_1 = 0.5   # Example coefficient for x
x = 10         # Example value for the predictor variable

log_odds_value = log_odds(beta_0, beta_1, x)

print(f"The log odds of orange versus apple is: {log_odds_value}")

def probability_orange(beta_0, beta_1, x):
    """
    Calculates the probability of orange given x, using the logistic function.
    """
    linear_combination = beta_0 + beta_1 * x
    return np.exp(linear_combination) / (1 + np.exp(linear_combination))

def odds_orange_vs_apple(beta_0, beta_1, x):
    """
    Calculates the odds of orange versus apple.
    """
    p_orange = probability_orange(beta_0, beta_1, x)
    return p_orange / (1 - p_orange)  # or np.exp(beta_0 + beta_1 * x)

# Verify that log(odds) == log_odds:
odds = odds_orange_vs_apple(beta_0, beta_1, x)
print(f"The odds of orange versus apple is: {odds}")
print(f"The log of the odds is: {np.log(odds)}")  # Should match log_odds_value
```

***(b) What is the log odds of orange versus apple in your friend’s model?***


您朋友的模型：

$P(Y = orange | X = x) = \frac{exp(\hat{α}_{orange0} + \hat{α}_{orange1}x)}{exp(\hat{α}_{orange0} + \hat{α}_{orange1}x) + exp(\hat{α}_{apple0} + \hat{α}_{apple1}x)}$

由於機率總和必須為 1，因此我們可以將蘋果的機率表示為：

$P(Y = apple | X = x) = \frac{exp(\hat{α}_{apple0} + \hat{α}_{apple1}x)}{exp(\hat{α}_{orange0} + \hat{α}_{orange1}x) + exp(\hat{α}_{apple0} + \hat{α}_{apple1}x)}$

計算勝算

橘子相對於蘋果的勝算是它們機率的比率：

$odds = \frac{P(Y = orange | X = x)}{P(Y = apple | X = x)}$

代入機率：

$odds = \frac{\frac{exp(\hat{α}_{orange0} + \hat{α}_{orange1}x)}{exp(\hat{α}_{orange0} + \hat{α}_{orange1}x) + exp(\hat{α}_{apple0} + \hat{α}_{apple1}x)}}{\frac{exp(\hat{α}_{apple0} + \hat{α}_{apple1}x)}{exp(\hat{α}_{orange0} + \hat{α}_{orange1}x) + exp(\hat{α}_{apple0} + \hat{α}_{apple1}x)}}$

請注意，分母相同並相互抵消：

$odds = \frac{exp(\hat{α}_{orange0} + \hat{α}_{orange1}x)}{exp(\hat{α}_{apple0} + \hat{α}_{apple1}x)}$

使用指數的性質，我們可以將其重寫為：

$odds = exp((\hat{α}_{orange0} + \hat{α}_{orange1}x) - (\hat{α}_{apple0} + \hat{α}_{apple1}x))$

計算對數勝算

現在，取勝算的自然對數：

$log\ odds = ln(exp((\hat{α}_{orange0} + \hat{α}_{orange1}x) - (\hat{α}_{apple0} + \hat{α}_{apple1}x)))$

自然對數和指數函數是反函數，因此它們相互抵消：

$log\ odds = (\hat{α}_{orange0} + \hat{α}_{orange1}x) - (\hat{α}_{apple0} + \hat{α}_{apple1}x)$

我們可以重新排列它：

$log\ odds = (\hat{α}_{orange0} - \hat{α}_{apple0}) + (\hat{α}_{orange1} - \hat{α}_{apple1})x$

答案

在您朋友的 softmax 模型中，橘子相對於蘋果的對數勝算是 $(\hat{α}_{orange0} - \hat{α}_{apple0}) + (\hat{α}_{orange1} - \hat{α}_{apple1})x$

***(c) Suppose that in your model, β_0 = 2 and β_1 = −1. What are the coefficient estimates in your friend’s model? Be as specific as possible.***


假設您的邏輯回歸模型具有 $β_0 = 2$ 且 $β_1 = -1$。

理解等價性

您的模型和您朋友的模型都在試圖表示相同的觀察值是橘子（或蘋果）的機率。 因此，兩個模型中的係數之間必須存在關係。 我們已經推導出：

您的模型的對數勝算：$β_0 + β_1x$
您朋友的模型的對數勝算：$(α_{orange0} - α_{apple0}) + (α_{orange1} - α_{apple1})x$

為了使這兩個表達式對於 x 的所有值都等效，係數必須相等：

$β_0 = α_{orange0} - α_{apple0}$
$β_1 = α_{orange1} - α_{apple1}$

使用給定的值

我們知道 $β_0 = 2$ 且 $β_1 = -1$。 所以：

$2 = α_{orange0} - α_{apple0}$
$-1 = α_{orange1} - α_{apple1}$

Softmax 約束

softmax 模型的一個關鍵特性是，它對於將常數添加到所有分數是不變的。 這意味著我們可以任意地將其中一個係數設置為固定值（通常為 0），以使模型可識別。 這是因為機率由分數之間的 *差異* 決定，而不是絕對值。

讓我們將蘋果係數設置為 0：

$α_{apple0} = 0$
$α_{apple1} = 0$

求解橘子係數

現在我們可以求解橘子係數：

$2 = α_{orange0} - 0 => α_{orange0} = 2$
$-1 = α_{orange1} - 0 => α_{orange1} = -1$

答案

因此，您朋友的模型的一個可能的係數估計值集合是：

$α_{orange0} = 2$
$α_{orange1} = -1$
$α_{apple0} = 0$
$α_{apple1} = 0$

**重要提示：** 這些 *不是* 唯一的可能解。 我們可以選擇將橘子係數設置為 0 並求解蘋果係數。 關鍵是係數之間的 *差異* 必須與您的模型的係數匹配。 此約束使我們能夠關聯兩個模型。 如果我們設置 alpha_orange0 = 0 且 alpha_orange1 = 0，則 alpha_apple0 = -2 且 alpha_apple1 = 1。

**總結：**

*   $α_{orange0} = 2, α_{orange1} = -1, α_{apple0} = 0, α_{apple1} = 0$ 是一個有效的解，而且可能是最簡單的。
*   $α_{orange0} = 0, α_{orange1} = 0, α_{apple0} = -2, α_{apple1} = 1$ 也是有效的。

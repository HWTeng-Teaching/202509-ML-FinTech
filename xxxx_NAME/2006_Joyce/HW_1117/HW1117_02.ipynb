9. In this exercise, we will predict the number of applications received using the other variables in the College data set.<br>
***(a) Split the data set into a training set and a test set.***

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# 1. 加 College 資料集
college = pd.read_csv('College.csv',index_col=0)

# 2. 定義預測變數 (X) 和響應變數 (Y)
# 根據題目，'Apps' 是我們要預測的收到的申請數量。
# 其他所有變數都將作為預測變數。
X = college.drop('Apps', axis=1) # 移除 'Apps' 列作為預測變數
Y = college['Apps']              # 'Apps' 列作為響應變數

# 檢查 X 和 Y 的形狀
print(f"\n預測變數 X 的維度: {X.shape}")
print(f"響應變數 Y 的維度: {Y.shape}")

# 3. 將資料集分成訓練集和測試集
# 我們將使用 70% 的資料作為訓練集，30% 的資料作為測試集。
# random_state 參數用於確保每次運行程式碼時，分割結果都相同，以保證可重現性。
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y,
    test_size=0.3,    # 測試集佔總資料的 30%
    random_state=42   # 設定隨機種子以確保結果可重現
)

# 4. 報告訓練集和測試集的維度
print(f"\n資料分割完成:")
print(f"訓練集預測變數 (X_train) 的維度: {X_train.shape}")
print(f"訓練集響應變數 (Y_train) 的維度: {Y_train.shape}")
print(f"測試集預測變數 (X_test) 的維度: {X_test.shape}")
print(f"測試集響應變數 (Y_test) 的維度: {Y_test.shape}")

# 顯示訓練集和測試集的部分數據以供驗證
print("\n訓練集預測變數 (X_train) 的前三行:")
print(X_train.head(3))
print("\n測試集響應變數 (Y_test) 的前三行:")
print(Y_test.head(3))
```

***(b) Fit a linear model using least squares on the training set, and report the test error obtained.***

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

# 2. 定義預測變數 (X) 和響應變數 (Y)
X = college.drop('Apps', axis=1)
Y = college['Apps']

# 由於 College 資料集中包含類別變數 (Private)，我們需要進行獨熱編碼 (One-Hot Encoding)
# LinearRegression 期望所有輸入都是數值型
X = pd.get_dummies(X, drop_first=True) # drop_first=True 避免多重共線性陷阱

# 3. 將資料集分成訓練集和測試集
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y,
    test_size=0.3,
    random_state=42
)

# --- (b) 在訓練集上擬合線性模型並報告測試誤差 ---

# 1. 創建線性迴歸模型實例
linear_model = LinearRegression()

# 2. 在訓練集上擬合模型
linear_model.fit(X_train, Y_train)

# 3. 使用擬合的模型在測試集上進行預測
Y_pred_test = linear_model.predict(X_test)

# 4. 計算測試集上的均方誤差 (MSE)
test_mse = mean_squared_error(Y_test, Y_pred_test)

# 5. 報告測試誤差
print(f"\n--- 線性迴歸模型結果 ---")
print(f"測試集上的均方誤差 (MSE): {test_mse:.2f}")

# 額外資訊：顯示模型係數
print("\n模型係數:")
# 結合特徵名稱和係數
coefficients = pd.Series(linear_model.coef_, index=X_train.columns)
print(coefficients)
print(f"\n截距 (Intercept): {linear_model.intercept_:.2f}")

# 也可以報告 R-squared，但題目只要求 MSE
# train_r2 = linear_model.score(X_train, Y_train)
# test_r2 = linear_model.score(X_test, Y_test)
# print(f"訓練集 R-squared: {train_r2:.4f}")
# print(f"測試集 R-squared: {test_r2:.4f}")
```

***(c) Fit a ridge regression model on the training set, with λ chosen by cross-validation. Report the test error obtained.***

```python
from sklearn.linear_model import LinearRegression, RidgeCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import numpy as np
# --- (c) 擬合 Ridge Regression 模型並報告測試誤差 ---

# 1. 標準化預測變數
# Ridge Regression 對變數尺度敏感，因此需要進行標準化
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train) # 在訓練集上擬合 StandardScaler 並轉換
X_test_scaled = scaler.transform(X_test)       # 使用訓練集的 StandardScaler 轉換測試集

# 將標準化後的數據轉換回 DataFrame，保留列名，以便模型解釋
X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_scaled_df = pd.DataFrame(X_test_scaled, columns=X_test.columns)


# 2. 創建 RidgeCV 模型實例並使用交叉驗證選擇最佳 alpha (λ)
# alphas 參數用於指定要測試的 λ 值範圍
# 我們通常在對數尺度上選擇一個範圍，因為 λ 的影響是非線性的
alphas = np.logspace(-4, 4, 200) # 測試從 0.0001 到 10000 的 200 個 alpha 值
                                 # 根據數據情況可以調整範圍

ridge_cv_model = RidgeCV(
    alphas=alphas,
    cv=10,             # 使用 10 折交叉驗證
    scoring='neg_mean_squared_error' # 選擇最小化 MSE 的 alpha
)

# 3. 在標準化後的訓練集上擬合模型
ridge_cv_model.fit(X_train_scaled_df, Y_train)

# 4. 獲取由交叉驗證選出的最佳 alpha (λ)
optimal_alpha_ridge = ridge_cv_model.alpha_

# 5. 使用最佳模型在標準化後的測試集上進行預測
Y_pred_test_ridge = ridge_cv_model.predict(X_test_scaled_df)

# 6. 計算測試集上的均方誤差 (MSE)
test_mse_ridge = mean_squared_error(Y_test, Y_pred_test_ridge)

# 7. 報告結果
print(f"\n--- 嶺迴歸 (Ridge Regression) 模型結果 ---")
print(f"通過交叉驗證選出的最佳 λ (alpha): {optimal_alpha_ridge:.6f}")
print(f"測試集上的均方誤差 (MSE): {test_mse_ridge:.2f}")

# 額外資訊：顯示最佳 Ridge 模型係數
print("\n最佳 Ridge 模型係數 (標準化數據):")
# 獲取最佳模型的係數。ridge_cv_model 已經是擬合好的最佳模型。
ridge_coefficients = pd.Series(ridge_cv_model.coef_, index=X_train.columns)
print(ridge_coefficients)
print(f"\n截距 (Intercept): {ridge_cv_model.intercept_:.2f}")

# --- (b) 線性迴歸模型結果 (重複報告用於比較) ---
print(f"\n--- (b) 線性迴歸模型結果 (用於比較) ---")
linear_model = LinearRegression()
linear_model.fit(X_train, Y_train)
Y_pred_test_linear = linear_model.predict(X_test)
test_mse_linear = mean_squared_error(Y_test, Y_pred_test_linear)
print(f"測試集上的均方誤差 (MSE): {test_mse_linear:.2f}")
```

***(d) Fit a lasso model on the training set, with λ chosen by cross-validation. Report the test error obtained, along with the number of non-zero coefficient estimates.***

```python
from sklearn.linear_model import LinearRegression, RidgeCV, LassoCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import numpy as np
import warnings

# 忽略來自 sklearn 可能關於收斂的警告
warnings.filterwarnings('ignore')

# --- (d) 擬合 Lasso 模型並報告測試誤差及非零係數數量 ---

# 1. 創建 LassoCV 模型實例並使用交叉驗證選擇最佳 alpha (λ)
# alphas 參數用於指定要測試的 λ 值範圍
# 我們通常在對數尺度上選擇一個範圍，因為 λ 的影響是非線性的
# max_iter 增加迭代次數以確保收斂
alphas_lasso = np.logspace(-4, 0, 100) # 測試從 0.0001 到 1 的 100 個 alpha 值
                                     # 通常 Lasso 的 alpha 範圍需要比 Ridge 更廣泛，且最大值不能太大，否則所有係數都為零

lasso_cv_model = LassoCV(
    alphas=alphas_lasso,
    cv=10,                      # 使用 10 折交叉驗證
    random_state=42,            # 確保交叉驗證分割的可重現性
    n_jobs=-1,                  # 使用所有可用處理器進行並行計算
    max_iter=10000              # 增加迭代次數以幫助收斂
)

# 2. 在標準化後的訓練集上擬合模型
lasso_cv_model.fit(X_train_scaled_df, Y_train)

# 3. 獲取由交叉驗證選出的最佳 alpha (λ)
optimal_alpha_lasso = lasso_cv_model.alpha_

# 4. 使用最佳模型在標準化後的測試集上進行預測
Y_pred_test_lasso = lasso_cv_model.predict(X_test_scaled_df)

# 5. 計算測試集上的均方誤差 (MSE)
test_mse_lasso = mean_squared_error(Y_test, Y_pred_test_lasso)

# 6. 報告結果
print(f"\n--- Lasso 模型結果 ---")
print(f"通過交叉驗證選出的最佳 λ (alpha): {optimal_alpha_lasso:.6f}")
print(f"測試集上的均方誤差 (MSE): {test_mse_lasso:.2f}")

# 7. 報告非零係數的數量
lasso_coefficients = lasso_cv_model.coef_
non_zero_coefficients_count = np.sum(lasso_coefficients != 0)
print(f"非零係數估計的數量: {non_zero_coefficients_count}")

# 額外資訊：顯示非零係數
print("\n非零係數估計 (標準化數據):")
coefficients_series = pd.Series(lasso_coefficients, index=X_train.columns)
for feature, coef in coefficients_series.items():
    if coef != 0:
        print(f"  {feature}: {coef:.4f}")
print(f"\n截距 (Intercept): {lasso_cv_model.intercept_:.2f}")


# --- 報告 (b) 和 (c) 的結果用於比較 ---
print(f"\n--- 各模型測試誤差比較 ---")

# (b) 線性迴歸模型結果
linear_model = LinearRegression()
linear_model.fit(X_train, Y_train)
Y_pred_test_linear = linear_model.predict(X_test)
test_mse_linear = mean_squared_error(Y_test, Y_pred_test_linear)
print(f"線性迴歸 (OLS) 測試 MSE: {test_mse_linear:.2f}")

# (c) 嶺迴歸模型結果
# 需要重新執行 RidgeCV 才能獲取其 MSE
alphas_ridge = np.logspace(-4, 4, 200)
ridge_cv_model_compare = RidgeCV(
    alphas=alphas_ridge,
    cv=10,
    scoring='neg_mean_squared_error'
)
ridge_cv_model_compare.fit(X_train_scaled_df, Y_train)
Y_pred_test_ridge = ridge_cv_model_compare.predict(X_test_scaled_df)
test_mse_ridge = mean_squared_error(Y_test, Y_pred_test_ridge)
print(f"嶺迴歸 (Ridge) 測試 MSE: {test_mse_ridge:.2f}")

print(f"Lasso 測試 MSE: {test_mse_lasso:.2f}")
```

#### **詳細解釋與討論：**

1.  **最佳 $\lambda$ (alpha) 值**：
    *   **數值**：您通過 10 折交叉驗證選擇出的最佳正則化參數 $\lambda$ (alpha) 約為 `0.000100`。
    *   **意義**：這是一個**非常小**的 $\lambda$ 值。在 Lasso 回歸中，$\lambda$ 控制著懲罰項的強度。當 $\lambda$ 值接近於零時，Lasso 的懲罰作用極弱，其行為會非常接近於普通的最小平方法 (OLS)。

2.  **測試誤差 (Test Error)**：
    *   **數值**：Lasso 模型在測試集上獲得的均方誤差 (MSE) 為 `1931803.39`。這個數值量化了模型預測的 `Apps` 值與實際 `Apps` 值之間的平均平方差異。

3.  **非零係數估計的數量**：
    *   **數值**：Lasso 模型中，非零係數估計的數量為 `17`。
    *   **意義**：這表示**所有 17 個預測變數**（包括 `Private_Yes` 獨熱編碼後）都被 Lasso 保留在模型中，沒有任何係數被完全壓縮為零。

4.  **模型係數 (標準化數據)**：
    *   **觀察**：報告中列出了所有 17 個非零係數的估計值，以及截距 `2964.50`。由於所有變數都被保留，這些係數與在微弱懲罰下的 OLS 係數非常相似。係數的絕對值大小反映了在標準化數據中該變數對 `Apps` 的相對重要性。例如，`Accept` 具有最大的正係數，`Enroll` 具有最大的負係數。

#### **與其他模型測試誤差的比較：**

*   **線性迴歸 (OLS) 測試 MSE**：`1931803.19`
*   **嶺迴歸 (Ridge) 測試 MSE**：`1931799.84`
*   **Lasso 測試 MSE**：`1931803.39`

*   **比較分析**：
    *   在您這次的實驗中，所有三個模型（OLS、Ridge、Lasso）在測試集上的 MSE 值都非常接近。Lasso 的 MSE 甚至略高於 OLS 和 Ridge。
    *   這再次強烈暗示在這個特定的 `College` 資料集和您所做的數據分割下，**模型的過擬合問題並不嚴重**。因此，正則化模型（Ridge 和 Lasso）的優勢未能充分體現，它們未能帶來顯著的預測性能提升。

#### **Lasso 在本案例中的表現總結：**

*   **未能實現特徵選擇**：Lasso 的一個主要優勢是能夠通過將不重要變數的係數壓縮為零來執行特徵選擇。然而，在您的結果中，**Lasso 並未將任何係數壓縮為零**，所有 17 個預測變數都被保留在模型中。
*   **原因分析**：
    1.  **極小的 $\lambda$ 值**：交叉驗證選擇的最佳 $\lambda$ 值 `0.000100` 非常小。這麼小的懲罰強度意味著模型幾乎沒有被要求犧牲任何係數來換取模型簡潔性。如果將任何係數壓縮為零會導致預測誤差增加，那麼 `LassoCV` 為了最小化預測誤差，就會傾向於不進行特徵選擇。
    2.  **所有變數可能都有貢獻**：這可能表明 `College` 資料集中的所有預測變數都對 `Apps` 的預測具有一定的貢獻，或者它們之間存在足夠的共線性，使得 Lasso 難以將任何一個變數的係數單獨壓縮為零而不顯著增加模型誤差。
    3.  **過擬合不嚴重**：所有模型預測性能的相似性也支持了這一點：由於沒有明顯的過擬合問題，正則化方法（Lasso）在性能提升和特徵選擇方面也就沒有太多發揮的空間。

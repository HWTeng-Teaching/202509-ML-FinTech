***(d) Now create a new function, Power3(), that actually returns the
result x ** a as a Python object, rather than simply printing it
to the screen. That is, if you store the value x ** a in an object
called result within your function, then you can simply return return this result, using the following line:***

**return** result

***Note that the line above should be the last line in your function, and it should be indented 4 spaces.***

```python
def Power3(x, a):
  """
  計算 x 的 a 次方並將結果作為 Python 物件返回。

  Args:
    x: 基礎數字。
    a: 指數。

  Returns:
    x 的 a 次方計算結果。
  """
  result = x ** a
  return result # 這行將計算出的結果返回

# --- 範例使用 ---

print("使用 Power3() 函數，其結果會被返回並可儲存到變數中:")

# 計算 2 的 3 次方，並將結果儲存到一個變數中
val1 = Power3(2, 3)
print(f"2 的 3 次方是: {val1}") # 這會印出 8

# 計算 5 的 2 次方，並將結果儲存到另一個變數中
val2 = Power3(5, 2)
print(f"5 的 2 次方是: {val2}") # 這會印出 25

# 您甚至可以直接在另一個操作中使用返回的值
val3 = Power3(10, 0.5) * 2
print(f"10 的 0.5 次方再乘以 2 是: {val3}") # 這會印出 6.324555320336759
```

***(e) Now using the Power3() function, create a plot of f(x) = x^2. The x-axis should display a range of integers from 1 to 10, and the y-axis should display x^2. Label the axes appropriately, and use an appropriate title for the figure. Consider displaying either the x-axis, the y-axis, or both on the log-scale. You can do this by using the ax.set_xscale() and ax.set_yscale() methods of the axes you are plotting to.***

```python
import matplotlib.pyplot as plt
import numpy as np

# 重新定義 Power3() 函數，以確保在獨立運行時可用
def Power3(x, a):
  """
  計算 x 的 a 次方並將結果作為 Python 物件返回。

  Args:
    x: 基礎數字。
    a: 指數。

  Returns:
    x 的 a 次方計算結果。
  """
  result = x ** a
  return result

# --- 1. 生成數據 ---
x_values = np.arange(1, 11) # 從 1 到 10 的整數
y_values = [Power3(x, 2) for x in x_values] # 使用 Power3 函數計算 x^2

# --- 2. 繪製圖表 (線性比例) ---
fig1, ax1 = plt.subplots(figsize=(8, 6))

ax1.plot(x_values, y_values, marker='o', linestyle='-', color='blue')
ax1.set_xlabel('x')
ax1.set_ylabel('f(x) = x^2')
ax1.set_title('Plot of f(x) = x^2 (Linear Scale)')
ax1.set_xticks(x_values) # 確保 x 軸顯示 1 到 10 的整數
ax1.grid(True)
plt.show()

# --- 3. 繪製圖表 (y 軸使用對數比例) ---
fig2, ax2 = plt.subplots(figsize=(8, 6))

ax2.plot(x_values, y_values, marker='o', linestyle='-', color='red')
ax2.set_xlabel('x')
ax2.set_ylabel('f(x) = x^2 (Log Scale)')
ax2.set_title('Plot of f(x) = x^2 (Y-axis on Log Scale)')
ax2.set_xticks(x_values) # 確保 x 軸顯示 1 到 10 的整數
ax2.set_yscale('log') # 將 y 軸設置為對數比例
ax2.grid(True, which="both", ls="-", alpha=0.6) # 在對數比例下顯示主要和次要網格線
plt.show()

# --- 說明 x 軸使用對數比例的情況 ---
# 雖然對於 1 到 10 這樣的範圍，x 軸使用對數比例可能不常見或不那麼直觀，
# 但為了展示 ax.set_xscale() 的用法，我們也可以這樣做：
fig3, ax3 = plt.subplots(figsize=(8, 6))
ax3.plot(x_values, y_values, marker='o', linestyle='-', color='green')
ax3.set_xlabel('x (Log Scale)')
ax3.set_ylabel('f(x) = x^2 (Linear Scale)')
ax3.set_title('Plot of f(x) = x^2 (X-axis on Log Scale)')
ax3.set_xscale('log') # 將 x 軸設置為對數比例
# 注意：在 log scale 下，xticks 的顯示可能會自動調整，
# 或者你需要手動指定如 ax3.set_xticks([1, 2, 5, 10]) 以便更好地視覺化。
ax3.grid(True, which="both", ls="-", alpha=0.6)
plt.show()
```

***(f) Create a function, PlotPower(), that allows you to create a plot of x against x**a for a fixed a and a sequence of values of x.***

>For instance, if you call

>PlotPower(np.arange(1, 11), 3)

***then a plot should be created with an x-axis taking on values 1,2,...,10, and a y-axis taking on values 1^3,2^3,...,10^3.***

```python
import matplotlib.pyplot as plt
import numpy as np

# 重新定義 Power3() 函數，以確保在獨立運行時可用
def Power3(x, a):
  """
  計算 x 的 a 次方並將結果作為 Python 物件返回。

  Args:
    x: 基礎數字。
    a: 指數。

  Returns:
    x 的 a 次方計算結果。
  """
  result = x ** a
  return result

def PlotPower(x_sequence, a):
  """
  繪製 x 與 x**a 的關係圖。

  Args:
    x_sequence: 一個 NumPy 陣列或列表，包含用於 x 軸的值。
    a: 一個數字，表示固定的指數。
  """
  # 使用 Power3() 函數計算對應的 y 值 (x**a)
  y_values = [Power3(x, a) for x in x_sequence]

  # 創建圖表
  plt.figure(figsize=(10, 6))
  plt.plot(x_sequence, y_values, marker='o', linestyle='-', color='purple')

  # 設置軸標籤和標題
  plt.xlabel('x')
  plt.ylabel(f'x^{a}') # 使用 f-string 讓標籤更動態
  plt.title(f'Plot of x vs. x^{a}')

  # 設置 x 軸刻度，以便在 x_sequence 是整數時顯示每個點
  if all(isinstance(val, (int, np.integer)) for val in x_sequence):
      plt.xticks(x_sequence)

  plt.grid(True)
  plt.show()

# --- 範例使用 ---

print("繪製 x 與 x^3 的關係圖 (x 從 1 到 10):")
PlotPower(np.arange(1, 11), 3)

print("\n繪製 x 與 x^0.5 的關係圖 (x 從 1 到 20):")
PlotPower(np.arange(1, 21), 0.5)

print("\n繪製 x 與 x^-2 的關係圖 (x 從 1 到 5):")
PlotPower(np.arange(1, 6), -2)
```

您遇到的 ValueError: Integers to negative integer powers are not allowed. 是一個在 Python 中當您嘗試將整數提高到負整數次冪時常見的問題。例如，2 ** -1 應該是 0.5，但 Python 為了避免在整數操作中進行隱式浮點數轉換，會拋出這個錯誤。

要解決這個問題，我們需要確保在 Power3() 函數中，如果底數 x 是整數且指數 a 是負整數時，將 x 轉換為浮點數再進行運算。

```python
import matplotlib.pyplot as plt
import numpy as np

def Power3(x, a):
  """
  計算 x 的 a 次方並將結果作為 Python 物件返回。

  Args:
    x: 基礎數字。
    a: 指數。

  Returns:
    x 的 a 次方計算結果。
  """
  # 檢查 x 是否為整數且 a 是否為負整數
  if isinstance(x, (int, np.integer)) and isinstance(a, (int, np.integer)) and a < 0:
    # 在執行冪運算之前將 x 轉換為浮點數
    result = float(x) ** a
  else:
    result = x ** a
  return result

def PlotPower(x_sequence, a):
  """
  繪製 x 與 x**a 的關係圖。

  Args:
    x_sequence: 一個 NumPy 陣列或列表，包含用於 x 軸的值。
    a: 一個數字，表示固定的指數。
  """
  # 使用 Power3() 函數計算對應的 y 值 (x**a)
  y_values = [Power3(x, a) for x in x_sequence]

  # 創建圖表
  plt.figure(figsize=(10, 6))
  plt.plot(x_sequence, y_values, marker='o', linestyle='-', color='purple')

  # 設置軸標籤和標題
  plt.xlabel('x')
  plt.ylabel(f'x^{a}') # 使用 f-string 讓標籤更動態
  plt.title(f'Plot of x vs. x^{a}')

  # 設置 x 軸刻度，以便在 x_sequence 是整數時顯示每個點
  # 判斷是否所有元素都是整數類型 (包括 numpy 內部的整數類型)
  if all(isinstance(val, (int, np.integer)) for val in x_sequence):
      plt.xticks(x_sequence)

  plt.grid(True)
  plt.show()

# --- 範例使用 ---

print("繪製 x 與 x^3 的關係圖 (x 從 1 到 10):")
PlotPower(np.arange(1, 11), 3)

print("\n繪製 x 與 x^0.5 的關係圖 (x 從 1 到 20):")
PlotPower(np.arange(1, 21), 0.5)

print("\n繪製 x 與 x^-2 的關係圖 (x 從 1 到 5):")
PlotPower(np.arange(1, 6), -2) # 現在這行可以正常運行了！
```

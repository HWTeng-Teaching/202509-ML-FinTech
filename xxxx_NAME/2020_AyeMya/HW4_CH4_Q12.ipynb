 12. Suppose that you wish to classify an observation X ∈ R into apples
 and oranges. You fit a logistic regression model and find that
 Pr(Y = orange|X = x) = exp(ˆβ0 + ˆβ1x)
 1 +exp(ˆβ0 + ˆβ1x).
 Your friend fits a logistic regression model to the same data using the
 softmax formulation in (4.13), and finds that
 Pr(Y = orange|X = x) =
 exp(ˆαorange0 + ˆαorange1x)
 exp(ˆαorange0 + ˆαorange1x) + exp(ˆαapple0 + ˆαapple1x) .
 (a) What is the log odds of orange versus apple in your model?
 (b) What is the log odds of orange versus apple in your friend’s
 model?
 (c) Suppose that in your model, ˆβ0 = 2 and ˆβ1 = −1. What are
 the coefficient estimates in your friend’s model? Be as specific
 as possible.
 (d) Now suppose that you and your friend fit the same two models
 on a different data set. This time, your friend gets the coefficient
 estimates ˆαorange0 = 1.2, ˆαorange1 = −2, ˆαorange0 = 3, ˆαorange1 =
 0.6. What are the coefficient estimates in your model?
 (e) Finally, suppose you apply both models from (d) to a data set
 with 2,000 test observations. What fraction of the time do you
 expect the predicted class labels from your model to agree with
 those from your friend’s model? Explain your answer.

(a) $$ \log(\text{odds}(\frac{\exp({\hat{\beta}_0 + \hat{\beta}_1X})}{1 + \exp({\hat{\beta}_0 + \hat{\beta}_1X})})) = \hat{\beta}_0 + \hat{\beta}_1X$$

(b) $$ \text{odds}(\frac{\hat{\text{Pr}}(Y=\text{orange}|X=x)}{\hat{\text{Pr}}(Y=\text{apple}|X=x)}) = \frac{\exp({\hat{\alpha}_{\text{orange0}}+ \hat{\alpha}_{\text{orange1}}x})}{\exp({\hat{\alpha}_{\text{apple0}}+ \hat{\alpha}_{\text{apple1}}x})}$$

$$ \log(\frac{\exp({\hat{\alpha}_{\text{orange0}}+ \hat{\alpha}_{\text{orange1}}x})}{\exp({\hat{\alpha}_{\text{apple0}}+ \hat{\alpha}_{\text{apple1}}x})}) = \hat{\alpha}_{\text{orange0}}+ \hat{\alpha}_{\text{orange1}}x - (\hat{\alpha}_{\text{apple0}}+ \hat{\alpha}_{\text{apple1}}x)$$

(c)
$$ \hat{\beta}_0 \approx \hat{\alpha}_{\text{orange0}} - \hat{\alpha}_{\text{apple0}}$$
$$ \hat{\beta}_0 \approx \hat{\alpha}_{\text{orange1}} - \hat{\alpha}_{\text{apple1}}$$

Constant term:
\[
\hat{\alpha}_{\text{orange}0} - \hat{\alpha}_{\text{apple}0} = \hat{\beta}_0 = 2.
\]

Slope on \( x \):
\[
\hat{\alpha}_{\text{orange}1} - \hat{\alpha}_{\text{apple}1} = \hat{\beta}_1 = -1.
\]

These equations do **not** have a unique solution, because in the softmax model we can add the same constant to both classes and predictions don't change. One clean (common) choice is to set the apple class as the baseline:

Take
\[
\hat{\alpha}_{\text{apple}0} = 0, \quad
\hat{\alpha}_{\text{apple}1} = 0.
\]

Then solve:
\[
\hat{\alpha}_{\text{orange}0} = 2, \quad
\hat{\alpha}_{\text{orange}1} = -1.
\]

**Answer (c): One valid parameterization**
- Apple:  
  \(\hat{\alpha}_{\text{apple}0} = 0,\quad \hat{\alpha}_{\text{apple}1} = 0\)

- Orange:  
  \(\hat{\alpha}_{\text{orange}0} = 2,\quad \hat{\alpha}_{\text{orange}1} = -1\)

Any pair of parameters that preserves the differences
\[
\hat{\alpha}_{\text{orange}0} - \hat{\alpha}_{\text{apple}0} = 2, \quad
\hat{\alpha}_{\text{orange}1} - \hat{\alpha}_{\text{apple}1} = -1
\]
is equivalent.

---

(d)
If we fit the same two models on a different data set, then
$$ \hat{\beta}_0 \approx \hat{\alpha}_{\text{orange0}} - \hat{\alpha}_{\text{apple0}} => \hat{\beta}_0 \approx 1.2 - 3 = -1.8$$
$$ \hat{\beta}_0 \approx \hat{\alpha}_{\text{orange1}} - \hat{\alpha}_{\text{apple1}} => \hat{\beta}_1 \approx -2 - 0.6 = -2.6$$

(e)
From (d), we saw that
\[
\hat{\beta}_0 + \hat{\beta}_1 x
= (\hat{\alpha}_{\text{orange}0} - \hat{\alpha}_{\text{apple}0})
+ (\hat{\alpha}_{\text{orange}1} - \hat{\alpha}_{\text{apple}1}) x
\]
for all \( x \).

That means:  
- both models produce the **same log-odds** of orange vs apple,  
- which means they produce the **same probability** of orange for every observation,  
- which means if you classify by "predict orange if probability > 0.5", you will make the **same class prediction**.

Therefore, for every test point, both models predict the same class label.

So on 2,000 test cases, they will agree on **all 2,000**.

**Answer (e):**
\[
\boxed{\text{They agree 100% of the time (fraction = 1.0).}}
\]

**Reason:** In the two-class case, binary logistic regression and two-class softmax are just different parameterizations of the same decision rule. They yield identical predicted probabilities and therefore identical class labels.

QUESTION C

Use the sm.OLS() function to perform a multiple linear regression with mpg as the response and all other variables except name as the predictors. Use the summarize() function to print the results. Comment on the output. For instance:

```python
import statsmodels.api as sm

X = MS(auto.columns.drop('mpg')).fit_transform(auto)
y = auto.iloc[:, 0]

model = sm.OLS(y, X)
results = model.fit()
print(results.summary())
```

C. i. Is there a relationship between the predictors and the response? Use the anova_lm() function from statsmodels to
answer this question.

```python
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm

# Build model using formula
model = ols('mpg ~ cylinders + displacement + horsepower + weight + acceleration + year + origin', data=auto).fit()

# Run ANOVA
anova_lm(model)
```

Insights:
1. 'cylinders' is statistically dominant, but it is likely collinear with other engine-related variables ('displacement', 'horsepower') (Redundant?)
2. 'year' is the most informative non-engine predictor. Newer cars will have more technology improvements
3. 'displacement', 'horsepower', and 'weight' all are significant and have relatively close F-scores. They could be overlapping, as they are all related to engine performance. We can check the interaction terms between the three variables later on
4. 'origin' is statistically significant, suggests regional design differences. Different regions will have different technology to reduce 'mpg'
5. 'acceleration' is not significant 

C. ii. Which predictors appear to have a statistically significant relationship to the response?

Looking from the OLS regression results, the variables in the column P>|t| (denoting p-values) with value less than 0.05 are significant variables. In this case, the variables with p-value (denoted by column P>|t|) less than 0.05 are 'displacement', 'weight', 'year', and 'origin'.

C. iii. What does the coefficient for the 'year' variable suggest?

The coefficient for the 'year' variable suggest that every 1 unit increase in 'year' variable will affect the value of 'mpg' response variable by 0.7508. That being said, if a car is newer by 1 year, then it will increase the distance the car is able to travel by 0.7508 per gallon of fuel. 

This positive correlation shows that newer cars will have better efficiency for fuel consumption.

QUESTION D

Produce some of diagnostic plots of the linear regression fit as described in the lab. Comment on any problems you see with the fit. Do the residual plots suggest any unusually large outliers? Does the leverage plot identify any observations with unusually high leverage?

```python
# fitted values (predicted values) v. residuals plot
from matplotlib.pyplot import subplots

ax = subplots(1, 1, figsize=(4, 4))[1]

ax.scatter(results.fittedvalues , results.resid)
ax.set_xlabel('Fitted value')
ax.set_ylabel('Residual')
ax.axhline(0, c='k', ls='--');
```

The analysis of the residual plot:
1. There seems to be a curve in the residual plot, suggesting non-linearity relationship between the predictors and the response variable.
2. There seems to be a cone shape or fanning out in the residual plot, suggesting heteroscedasticity in the model.
3. The fitted values with residual value above 10 seems to be outliers.

```python
# Creating leverage plot and leverage-studentized residuals plot to see high-leverage observation

infl = results.get_influence()
leverage = infl.hat_matrix_diag
student_resid = infl.resid_studentized_external

axs = subplots(1, 2, figsize=(6, 3))[1]
axs[0].scatter(np.arange(X.shape[0]), infl.hat_matrix_diag)
axs[0].set_xlabel('Index')
axs[0].set_ylabel('Leverage')

# Annotate points with high leverage or residuals
for i in range(len(leverage)):
    if leverage[i] > 0.1 or abs(student_resid[i]) < 0:  # Customize thresholds
        axs[0].annotate(str(i), (i, leverage[i]), fontsize=9, color='red')
        axs[1].annotate(str(i), (leverage[i], student_resid[i]), fontsize=9, color='red')
        
axs[1].scatter(infl.hat_matrix_diag, infl.resid_studentized_external)
axs[1].set_xlabel('Leverage')
axs[1].set_ylabel('Studentized Residuals')

plt.tight_layout()
        
np.argmax(infl.hat_matrix_diag)
```

The analysis of the leverage plot:
There is an observation detected with high leverage value (observation number 13). This shows that the regression model might be affected
1. The regression coefficients might be biased
2. It may cause the model to underestimate residuals, leading to undetected (or masked) outliers
3. It may inflate Rsquare and give misleading fit, leading to false sense of accuracy
4. It causes unstable predictions
5. Standard errors and p-values can be distorted, leading to incorrect inferences on significance level

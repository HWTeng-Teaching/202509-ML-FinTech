```python
import numpy as np
import matplotlib.pyplot as plt

# Part (a): Power function that prints 2^3
print("=== Part (a) ===")
def Power():
    """Prints the result of raising 2 to the 3rd power"""
    result = 2 ** 3
    print(f"2^3 = {result}")

# Test the function
Power()

# Part (b): Power2 function that takes any two numbers
print("\n=== Part (b) ===")
def Power2(x, a):
    """Prints the result of raising x to the power a"""
    result = x ** a
    print(f"{x}^{a} = {result}")

# Test with the given example: 3^8 = 6561
Power2(3, 8)

# Part (c): Compute various powers using Power2
print("\n=== Part (c) ===")
Power2(10, 3)   # 10^3
Power2(8, 17)   # 8^17
Power2(131, 3)  # 131^3

# Part (d): Power3 function that returns the result
print("\n=== Part (d) ===")
def Power3(x, a):
    """Returns the result of raising x to the power a"""
    result = x ** a
    return result

# Test Power3 function
test_result = Power3(3, 4)
print(f"Power3(3, 4) returns: {test_result}")

# Part (e): Plot f(x) = x^2 using Power3 function
print("\n=== Part (e) ===")

# Create x values from 1 to 10
x_values = np.arange(1, 11)
# Compute y values using Power3 function
y_values = [Power3(x, 2) for x in x_values]

# Create the plot
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Plot 1: Regular scale
axes[0, 0].plot(x_values, y_values, 'bo-', linewidth=2, markersize=6)
axes[0, 0].set_xlabel('x')
axes[0, 0].set_ylabel('f(x) = x²')
axes[0, 0].set_title('f(x) = x² (Regular Scale)')
axes[0, 0].grid(True, alpha=0.3)

# Plot 2: Log scale on y-axis
axes[0, 1].plot(x_values, y_values, 'ro-', linewidth=2, markersize=6)
axes[0, 1].set_xlabel('x')
axes[0, 1].set_ylabel('f(x) = x²')
axes[0, 1].set_title('f(x) = x² (Log Y-scale)')
axes[0, 1].set_yscale('log')
axes[0, 1].grid(True, alpha=0.3)

# Plot 3: Log scale on x-axis
axes[1, 0].plot(x_values, y_values, 'go-', linewidth=2, markersize=6)
axes[1, 0].set_xlabel('x')
axes[1, 0].set_ylabel('f(x) = x²')
axes[1, 0].set_title('f(x) = x² (Log X-scale)')
axes[1, 0].set_xscale('log')
axes[1, 0].grid(True, alpha=0.3)

# Plot 4: Log scale on both axes
axes[1, 1].plot(x_values, y_values, 'mo-', linewidth=2, markersize=6)
axes[1, 1].set_xlabel('x')
axes[1, 1].set_ylabel('f(x) = x²')
axes[1, 1].set_title('f(x) = x² (Log-Log Scale)')
axes[1, 1].set_xscale('log')
axes[1, 1].set_yscale('log')
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Part (f): PlotPower function - FIXED VERSION
print("\n=== Part (f) ===")
def PlotPower(x_range, a):
    """
    Creates a plot of x against x^a for a fixed a and sequence of x values

    Parameters:
    x_range: array-like, sequence of x values
    a: exponent value
    """
    # Convert to float array to handle negative exponents properly
    x_range_float = np.array(x_range, dtype=float)

    # Compute y values - handle case where x=0 and a is negative
    y_values = []
    for x in x_range_float:
        if x == 0 and a < 0:
            y_values.append(np.inf)  # Handle division by zero
        else:
            y_values.append(x ** a)

    y_values = np.array(y_values)

    # Create the plot
    plt.figure(figsize=(10, 6))
    plt.plot(x_range_float, y_values, 'b-', linewidth=2, marker='o', markersize=6)
    plt.xlabel('x')
    plt.ylabel(f'f(x) = x^{a}')
    plt.title(f'Plot of f(x) = x^{a}')
    plt.grid(True, alpha=0.3)

    # Adjust y-axis limits for better visualization, especially for negative exponents
    if a < 0:
        # For negative exponents, use log scale or limit the y-axis
        plt.ylim(bottom=0, top=max(y_values[y_values < np.inf]) * 1.1)

    # Show the plot
    plt.show()

    # Also print the values in a table
    print(f"Values for f(x) = x^{a}:")
    print("x\tf(x)")
    print("-" * 20)
    for x, y in zip(x_range_float, y_values):
        if y == np.inf:
            print(f"{x}\tundefined (division by zero)")
        else:
            print(f"{x}\t{y:.6f}")

# Test PlotPower function with the given example
print("Testing PlotPower with x = 1 to 10, a = 3:")
PlotPower(np.arange(1, 11), 3)

# Additional examples to demonstrate the flexibility of PlotPower
print("\n=== Additional Examples ===")

# Example 1: Different exponent
print("\nExample 1: f(x) = x^4 for x = 1 to 5")
PlotPower(np.arange(1, 6), 4)

# Example 2: Larger range
print("\nExample 2: f(x) = x^1.5 for x = 1 to 15")
PlotPower(np.arange(1, 16), 1.5)

# Example 3: Fractional exponent
print("\nExample 3: f(x) = x^0.5 (square root) for x = 1 to 20")
PlotPower(np.arange(1, 21), 0.5)

# Example 4: Negative exponent - FIXED
print("\nExample 4: f(x) = x^(-1) for x = 1 to 10")
PlotPower(np.arange(1, 11), -1)

# Example 5: Another negative exponent with different range
print("\nExample 5: f(x) = x^(-2) for x = 1 to 8")
PlotPower(np.arange(1, 9), -2)

# Example 6: Including zero with positive exponent
print("\nExample 6: f(x) = x^2 for x = 0 to 5")
PlotPower(np.arange(0, 6), 2)

# Demonstrate all functions working together
print("\n=== Final Demonstration ===")
print("Demonstrating all functions working together:")

# Use Power()
print("\n1. Using Power():")
Power()

# Use Power2()
print("\n2. Using Power2(5, 4):")
Power2(5, 4)

# Use Power3() and store result
print("\n3. Using Power3(7, 3) and storing result:")
result = Power3(7, 3)
print(f"Stored result: {result}")

# Use the stored result in calculations
print(f"Using the result in calculation: {result} * 2 = {result * 2}")

# Use PlotPower() with custom range
print("\n4. Using PlotPower() with custom parameters:")
custom_x = [1, 2, 3, 4, 5, 6, 7, 8]
PlotPower(custom_x, 2)

# Additional demonstration with decimal exponents
print("\n5. Using PlotPower() with decimal exponents:")
PlotPower([1, 2, 3, 4, 5], 0.7)

print("\n=== Function Definitions Summary ===")
print("""
Power(): Prints 2^3
Power2(x, a): Prints x^a
Power3(x, a): Returns x^a as a value
PlotPower(x_range, a): Creates a plot of x^a for given x values
""")

# Show the mathematical relationship for negative exponents
print("\n=== Mathematical Note ===")
print("For negative exponents like x^(-1):")
print("x^(-1) = 1/x")
print("x^(-2) = 1/(x^2)")
print("This is why the values decrease as x increases for negative exponents.")
```

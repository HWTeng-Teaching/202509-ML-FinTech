(a) Based on this data set, provide an estimate for the population
mean of medv. Call this estimateˆ
µ.

```python
import numpy as np
import pandas as pd


Boston = pd.read_csv('Boston.csv')

mu_hat = Boston['medv'].mean()
print(mu_hat)
```

(b) Provide an estimate of the standard error of
ˆµ. Interpret this
result.
Hint: We can compute the standard error of the sample mean by
dividing the sample standard deviation by the square root of the
number of observations.

```python
medv = Boston['medv']
n = len(medv)
s = medv.std(ddof=1)
se_mu = s / np.sqrt(n)

mu_hat = medv.mean()

print(se_mu)
```

(c) Now estimate the standard error of ˆµ using the bootstrap. How
does this compare to your answer from (b)?

```python
import numpy as np

medv = Boston['medv'].values
n = len(medv)

rng = np.random.default_rng(1)
B = 1000

boot_means = np.empty(B)

for b in range(B):
    sample_idx = rng.integers(0, n, size=n)
    sample = medv[sample_idx]
    boot_means[b] = sample.mean()

se_boot = boot_means.std(ddof=1)
print(se_boot)
```

larger than (b)

```python
ci_boot_lower = mu_hat - 2 * se_boot
ci_boot_upper = mu_hat + 2 * se_boot
ci_boot = (ci_boot_lower, ci_boot_upper)
print(ci_boot)
```

(f) We now would like to estimate the standard error of ˆµmed. Unfortunately, there is no simple formula for computing the standard
error of the median. Instead, estimate the standard error of the
median using the bootstrap. Comment on your findings.

```python
import numpy as np

medv = Boston['medv'].values
n = len(medv)

mu_med_hat = np.median(medv)

rng = np.random.default_rng(1)
B = 1000

boot_medians = np.empty(B)

for b in range(B):
    sample_idx = rng.integers(0, n, size=n)
    sample = medv[sample_idx]
    boot_medians[b] = np.median(sample)

se_med_boot = boot_medians.std(ddof=1)
print(mu_med_hat)
print(se_med_boot)
```

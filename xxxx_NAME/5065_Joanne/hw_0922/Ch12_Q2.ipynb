```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial.distance import squareform
from scipy.cluster.hierarchy import linkage, dendrogram

M = np.array([
    [0.0, 0.3, 0.4, 0.7],
    [0.3, 0.0, 0.5, 0.8],
    [0.4, 0.5, 0.0, 0.45],
    [0.7, 0.8, 0.45, 0.0]
], dtype=float)

# covert into condensed vector，then do complete linkage
Z = linkage(squareform(M), method="complete")

plt.figure(figsize=(5,4))
dendrogram(Z, labels=["O1","O2","O3","O4"])
plt.title("Hierarchical Clustering Dendrogram (Complete Linkage)")
plt.xlabel("Observations")
plt.ylabel("Dissimilarity")
plt.axhline(0.30, ls="--", lw=0.8)
plt.axhline(0.45, ls="--", lw=0.8)
plt.axhline(0.80, ls="--", lw=0.8)
plt.tight_layout()
plt.show()
```

(b) Repeat (a), this time using single linkage clustering.

```python
# Single linkage
Z = linkage(squareform(M), method="single")

plt.figure(figsize=(5,4))
dendrogram(Z, labels=["O1","O2","O3","O4"])
plt.title("Hierarchical Clustering Dendrogram (Single Linkage)")
plt.xlabel("Observations")
plt.ylabel("Dissimilarity")
for h in (0.30, 0.40, 0.45):
    plt.axhline(h, ls="--", lw=0.8)
plt.tight_layout()
plt.show()
```

(e) It is mentioned in this chapter that at each fusion in the den-
drogram, the position of the two clusters being fused can be
swapped without changing the meaning of the dendrogram. Draw
a dendrogram that is equivalent to the dendrogram in (a), for
which two or more of the leaves are repositioned, but for which
the meaning of the dendrogram is the same.

```python
labels = ["O1","O2","O3","O4"]

Z = linkage(squareform(M), method="complete")  # complete linkage
Z_eq = Z.copy()
Z_eq[:, [0,1]] = Z_eq[:, [1,0]]               # swap the left and right

plt.figure(figsize=(9,4))
plt.subplot(1,2,2); dendrogram(Z_eq, labels=labels); plt.title("Equivalent (count sort)")
plt.tight_layout(); plt.show()
```

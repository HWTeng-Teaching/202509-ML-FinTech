# 1110HW_C05_Q09 - Boston Housing Analysis

This notebook restates each sub-question in English and answers it in its own code cell by working with the Boston housing data (`boston.csv`) from the ISLP library.


```python
import numpy as np
import pandas as pd

BOSTON_PATH = "boston.csv"
boston = pd.read_csv(BOSTON_PATH)
medv = boston["medv"]
medv_values = medv.to_numpy()
n_obs = medv_values.shape[0]
BOOTSTRAP_REPS = 5000

def bootstrap_statistic(values, func, n_boot=BOOTSTRAP_REPS, random_state=None):
    rng = np.random.default_rng(random_state)
    n = values.shape[0]
    stats = np.empty(n_boot)
    for i in range(n_boot):
        resample = rng.choice(values, size=n, replace=True)
        stats[i] = func(resample)
    return stats
```

**(a) Based on this data set, provide an estimate for the population mean of `medv`. Call this estimate $\hat{\mu}$.**


```python
mu_hat = float(medv_values.mean())
print(f"Number of Boston census tracts: {n_obs}")
print(f"Sample mean of medv (mu_hat): {mu_hat:.3f} thousand dollars")
print("This sample mean is the plug-in estimate for the population mean of medv.")
```

**(b) Provide an estimate of the standard error of $\hat{\mu}$. Interpret this result.**

Hint: We can compute the standard error of the sample mean by dividing the sample standard deviation by the square root of the number of observations.


```python
sample_sd = float(medv_values.std(ddof=1))
se_mu = sample_sd / np.sqrt(n_obs)
print(f"Sample standard deviation of medv: {sample_sd:.3f}")
print(f"Standard error of mu_hat: {se_mu:.3f}")
print("Interpretation: across repeated samples of 506 tracts, the sample mean would typically vary by about this many thousand dollars.")
```

**(c) Now estimate the standard error of $\hat{\mu}$ using the bootstrap. How does this compare to your answer from (b)?**


```python
boot_means = bootstrap_statistic(medv_values, np.mean, random_state=0)
boot_se_mu = float(boot_means.std(ddof=1))
print(f"Bootstrap standard error of mu_hat: {boot_se_mu:.3f}")
print(f"Analytic standard error from part (b): {se_mu:.3f}")
print(f"Difference (bootstrap - analytic): {boot_se_mu - se_mu:.6f}")
```

**(d) Based on your bootstrap estimate from (c), provide a 95% confidence interval for the mean of `medv`. Compare it to the results obtained by using `Boston['medv'].std()` and the two standard error rule (3.9).**

Hint: You can approximate a 95% confidence interval using the formula [$\hat{\mu} - 2SE(\hat{\mu}),\, \hat{\mu} + 2SE(\hat{\mu})$].


```python
ci_boot_rule = (mu_hat - 2 * boot_se_mu, mu_hat + 2 * boot_se_mu)
ci_analytic = (mu_hat - 2 * se_mu, mu_hat + 2 * se_mu)
ci_percentile = np.percentile(boot_means, [2.5, 97.5])
print(f"Bootstrap 95% CI via 2*SE rule: ({ci_boot_rule[0]:.3f}, {ci_boot_rule[1]:.3f}) thousand dollars")
print(f"Analytic 95% CI via 2*SE rule: ({ci_analytic[0]:.3f}, {ci_analytic[1]:.3f}) thousand dollars")
print(f"Bootstrap percentile 95% CI: ({ci_percentile[0]:.3f}, {ci_percentile[1]:.3f}) thousand dollars")
print("Both approaches yield almost the same interval because the sampling distribution of the mean is close to normal.")
```

**(e) Based on this data set, provide an estimate, $\hat{\mu}_{\text{med}}$, for the median value of `medv` in the population.**


```python
median_hat = float(np.median(medv_values))
print(f"Sample median of medv (mu_med): {median_hat:.3f} thousand dollars")
```

**(f) We now would like to estimate the standard error of $\hat{\mu}_{\text{med}}$. Unfortunately, there is no simple formula for computing the standard error of the median. Instead, estimate the standard error of the median using the bootstrap. Comment on your findings.**


```python
boot_medians = bootstrap_statistic(medv_values, np.median, random_state=1)
boot_se_median = float(boot_medians.std(ddof=1))
print(f"Bootstrap standard error of mu_med: {boot_se_median:.3f}")
print("The sampling variability of the median is slightly smaller than that of the mean, reflecting the median's robustness to extreme values.")
```

**(g) Based on this data set, provide an estimate for the tenth percentile of `medv` in Boston census tracts. Call this quantity $\hat{\mu}_{0.1}$. (You can use the `np.percentile()` function.)**


```python
mu_p10 = float(np.percentile(medv_values, 10))
print(f"Tenth percentile estimate (mu_0.1): {mu_p10:.3f} thousand dollars")
```

**(h) Use the bootstrap to estimate the standard error of $\hat{\mu}_{0.1}$. Comment on your findings.**


```python
boot_p10 = bootstrap_statistic(medv_values, lambda sample: np.percentile(sample, 10), random_state=2)
boot_se_p10 = float(boot_p10.std(ddof=1))
print(f"Bootstrap standard error of mu_0.1: {boot_se_p10:.3f}")
print("Lower-tail estimates are less stable; the tenth percentile varies a bit more than the mean or median across bootstrap samples.")
```

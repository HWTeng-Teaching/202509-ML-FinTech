# 1020HW C04 Q06 — Logistic Regression
We collect data for a statistics class with predictors `X1 = hours studied`, `X2 = undergrad GPA`, and response `Y = receive an A`.
A logistic regression is fit with estimated coefficients: $\beta_0 = -6$, $\beta_1 = 0.05$, $\beta_2 = 1$.\n
(a) Estimate the probability that a student who studies 40 hours and has an undergrad GPA of 3.5 gets an A.\n
(b) How many hours would that student (GPA=3.5) need to study to have a 50% chance of an A?

```python
import math

# Coefficients
b0, b1, b2 = -6.0, 0.05, 1.0

def logistic(z):
    return 1.0 / (1.0 + math.exp(-z))

# (a) Probability for 40 hours, GPA 3.5
x1, x2 = 40.0, 3.5
z = b0 + b1 * x1 + b2 * x2
p = logistic(z)
print(f'(a) z = {z:.3f}, probability = {p:.6f}')

# (b) Hours needed for 50% chance (p=0.5 => z=0) with GPA fixed at 3.5
x2_fixed = 3.5
x1_needed = (-b0 - b2 * x2_fixed) / b1
print(f'(b) hours for 50% chance (GPA=3.5): {x1_needed:.1f} hours')
```

Answer:
- (a) z = -0.5, probability ≈ 0.3775.
- (b) Solve 0 = -6 + 0.05·hours + 3.5 ⇒ hours = 50.

# 1020HW C04 Q12 - Logistic vs Softmax (Binary)
This question compares binary logistic regression and two-class softmax.


## (a) Log-odds in your model
What is the log-odds of orange versus apple in your model?


### Answer (a)
log[p(orange|x)/p(apple|x)] = b0 + b1 x.


## (b) Log-odds in friend's softmax model
What is the log-odds of orange versus apple in your friend's model?


### Answer (b)
log[p(orange|x)/p(apple|x)] = (a_o0 - a_a0) + (a_o1 - a_a1) x.


## (c) Mapping when b0=2, b1=-1
Suppose in your model b0=2 and b1=-1. What are the coefficient estimates in your friend's model? Be as specific as possible.


### Answer (c)
Only differences are identified: b_j = a_oj - a_aj. One valid choice: a_o0=2, a_o1=-1, a_a0=0, a_a1=0. General family: a_o0=2+c, a_a0=c; a_o1=-1+d, a_a1=d.


## (d) Convert friend's estimates to b
Friend gets a_orange0=1.2, a_orange1=-2, a_apple0=3, a_apple1=0.6. What are the coefficients in your model?


### Answer (d)
b0 = 1.2 - 3 = -1.8; b1 = -2 - 0.6 = -2.6.


## (e) Agreement
If both models from (d) are applied to 2,000 test observations, what fraction of predicted labels agree? Explain.


### Answer (e)
100%. With b_j = a_oj - a_aj, two-class softmax equals logistic for all x.


```python
# Optional numeric check for (d)/(e)
import math
a_o0, a_o1 = 1.2, -2.0
a_a0, a_a1 = 3.0, 0.6
b0 = a_o0 - a_a0
b1 = a_o1 - a_a1
def p_softmax(x):
    so = math.exp(a_o0 + a_o1*x)
    sa = math.exp(a_a0 + a_a1*x)
    return so/(so+sa)
def p_logistic(x):
    z = b0 + b1*x
    return math.exp(z)/(1+math.exp(z))
xs = [-2,-1,0,0.5,1,2]
diffs = [abs(p_softmax(x)-p_logistic(x)) for x in xs]
b0, b1, diffs
```

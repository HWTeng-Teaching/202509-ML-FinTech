# 1020HW C04 Q09 — Odds
This problem has to do with odds.\n
**(a)** On average, what fraction of people with an odds of 0.37 of defaulting on their credit card payment will in fact default?\n
**(b)** Suppose that an individual has a 16% chance of defaulting on her credit card payment. What are the odds that she will default?

## Formulas
- If odds are `o`, probability is `p = o / (1 + o)`.
- If probability is `p`, odds are `o = p / (1 - p)`.

```python
# Compute answers
odds_a = 0.37
p_a = odds_a / (1 + odds_a)  # fraction who default given odds

p_b = 0.16
odds_b = p_b / (1 - p_b)     # odds given probability

print(f'(a) Probability of default given odds 0.37: {p_a:.6f} ({p_a:.2%})')
print(f'(b) Odds given p=0.16: {odds_b:.6f} (~{odds_b:.2f}:1)')
```

## Answers
- (a) `p = 0.37 / (1 + 0.37) ≈ 0.2701` → about **27.01%**.
- (b) `odds = 0.16 / (1 − 0.16) = 0.16 / 0.84 ≈ 0.190476 ` → odds **0.1905**.

```python
import numpy as np
import pandas as pd
path = r"C:\Users\User-NB\OneDrive\Desktop\ml&fintech\1110\Boston.csv"
Boston = pd.read_csv(path)

Boston.head()

medv = Boston["medv"].values  
n = len(medv)
n
```

 (a) Based on this data set, provide an estimate for the population
 mean of medv. Call this estimate ˆµ

```python
mu_hat = medv.mean()
mu_hat
```

 (b) Provide an estimate of the standard error of ˆµ. Interpret this
 result.
 Hint: We can compute the standard error of the sample mean by
 dividing the sample standard deviation by the square root of the
 number of observations.

```python
# 樣本標準差：ddof=1 代表除以 (n-1)
s = medv.std(ddof=1)
se_mu = s / np.sqrt(n)
s, se_mu
```

 (c) Now estimate the standard error of ˆµ using the bootstrap. How
 does this compare to your answer from (b)?

```python
rng = np.random.default_rng(123)  # 設 seed 讓結果可重現
B = 10_000                        # bootstrap 次數

boot_means = np.empty(B)

for b in range(B):
    sample = rng.choice(medv, size=n, replace=True)
    boot_means[b] = sample.mean()

se_mu_boot = boot_means.std(ddof=1)
se_mu_boot
```

 (d) Based on your bootstrap estimate from (c), provide a 95% con
f
 idence interval for the mean of medv. Compare it to the results
 obtained by using Boston['medv'].std() and the two standard
 error rule (3.9).
 Hint: You can approximate a 95% confidence interval using the
 formula [ˆµ 2SE(ˆµ), ˆµ + 2SE(ˆµ)].

```python
ci_boot_low = mu_hat - 2 * se_mu_boot
ci_boot_high = mu_hat + 2 * se_mu_boot
(ci_boot_low, ci_boot_high)
```

```python
ci_theory_low = mu_hat - 2 * se_mu
ci_theory_high = mu_hat + 2 * se_mu
(ci_theory_low, ci_theory_high)
```

Bootstrap 與常態近似的結果高度一致。

 (f) We now would like to estimate the standard error of ˆµmed. Unfortunately, there is no simple formula for computing the standard
 error of the median. Instead, estimate the standard error of the
 median using the bootstrap. Comment on your findings.

```python
rng = np.random.default_rng(456)  # 換個 seed
B = 10_000
boot_medians = np.empty(B)

for b in range(B):
    sample = rng.choice(medv, size=n, replace=True)
    boot_medians[b] = np.median(sample)

se_med_boot = boot_medians.std(ddof=1)
se_med_boot
```

 (h) Use the bootstrap to estimate the standard error of ˆµ0.1. Com
ment on your findings.

```python
rng = np.random.default_rng(789)
B = 10_000
boot_p10 = np.empty(B)

for b in range(B):
    sample = rng.choice(medv, size=n, replace=True)
    boot_p10[b] = np.percentile(sample, 10)

se_p10_boot = boot_p10.std(ddof=1)
se_p10_boot
```

這個標準誤告訴我們：在相同母體下重複抽樣，第 10 百分位數估計值會有多大的變動。

通常第 10 百分位這種「尾端」統計量，比平均數更不穩定一點，Bootstrap 可以捕捉到這種不穩定性

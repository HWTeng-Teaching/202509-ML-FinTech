15. This problem involves the Boston data set, which we saw in the lab
for this chapter. We will now try to predict per capita crime rate
using the other variables in this data set. In other words, per capita
crime rate is the response, and the other variables are the predictors.

(a) For each predictor, fit a simple linear regression model to predict
the response. Describe your results. In which of the models is
there a statistically significant association between the predictor
and the response? Create some plots to back up your assertions.

```python
import pandas as pd

boston = pd.read_csv(r"C:\Users\User-NB\OneDrive\Desktop\ml&fintech\1013\Boston.csv")

if "Unnamed: 0" in boston.columns:
    boston = boston.drop(columns=["Unnamed: 0"])
    
print(boston.head())
print(boston.columns)
```

```python
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt

response = 'crim' if 'crim' in boston.columns else 'CRIM'
predictors = [c for c in boston.columns if c.lower() != 'crim']

simple_results = {}

for var in predictors:
    formula = f"{response} ~ {var}"
    model = smf.ols(formula=formula, data=boston).fit()
    simple_results[var] = model.params[var]
    pval = model.pvalues[var]
    print(f"{var:>8}: coef={model.params[var]:.4f}, p={pval:.4g}")

    plt.figure()
    plt.scatter(boston[var], boston[response], alpha=0.6)
    plt.xlabel(var); plt.ylabel(response)
    plt.title(f"{response} vs {var}")
    x = boston[var]
    yhat = model.params[0] + model.params[1]*x
    plt.plot(x, yhat, color='red')
    plt.show()
```

多數變數（例如 indus, nox, tax, rad, lstat）與 crim 顯著相關 (p < 0.05)。

dis（距離工作中心）與 crim 呈負相關；rad、tax 呈正相關。

chas（是否臨河）通常不顯著。

某些變數的散點圖顯示非線性趨勢（例如 dis, nox）。

(b) Fit a multiple regression model to predict the response using
all of the predictors. Describe your results. For which predictors
can we reject the null hypothesis H0 : βj = 0?

```python
formula_all = f"{response} ~ " + " + ".join(predictors)
multi_model = smf.ols(formula=formula_all, data=boston).fit()

print(multi_model.summary())
```

F 檢定 p-value < 0.001，顯示整體模型顯著。

顯著的變數有 zn, dis, rad, black, lstat。

rad 係數為正：高速公路可達性越高、犯罪率越高。

dis 為負：距市中心越遠、犯罪率越低。

模型 R² 約 0.45~0.50，解釋力中等。

(c) How do your results from (a) compare to your results from (b)?
Create a plot displaying the univariate regression coefficients
from (a) on the x-axis, and the multiple regression coefficients
from (b) on the y-axis. That is, each predictor is displayed as a
single point in the plot. Its coefficient in a simple linear regression model is shown on the x-axis, and its coefficient estimate
in the multiple linear regression model is shown on the y-axis.

```python
import matplotlib.pyplot as plt

simple_coef = pd.Series(simple_results)
multi_coef = multi_model.params.drop("Intercept")

plt.figure(figsize=(6,6))
plt.scatter(simple_coef, multi_coef.loc[simple_coef.index])
plt.axhline(0, color='gray', linestyle='--')
plt.axvline(0, color='gray', linestyle='--')
plt.xlabel("Simple Regression Coefficient")
plt.ylabel("Multiple Regression Coefficient")
plt.title("(c) Simple vs Multiple Regression Coefficients")

# 標註變數名稱
for var in simple_coef.index:
    plt.text(simple_coef[var], multi_coef[var], var, fontsize=9)
plt.show()
```

圖中大部分變數的單變數與多變數係數都接近 0，表示其對犯罪率的影響在控制其他變數後不再顯著。
唯一明顯偏離的是 NOX，在單變數下對犯罪率呈強烈正相關（x 軸很大），
但在多變數模型中係數變為負（y 軸 < 0）。
這代表 nox 與其他變數（例如 indus, dis）高度相關，
在控制這些變數後，其獨立效應方向反轉。

(d) Non-Linear Associations

Is there evidence of non-linear association between any of the predictors and the response?  
To answer this question, for each predictor $X$, fit a model of the form:

$$
Y = \beta_0 + \beta_1 X + \beta_2 X^2 + \beta_3 X^3 + \epsilon
$$


```python
nonlinear_results = {}

for var in predictors:
    formula = f"{response} ~ {var} + I({var}**2) + I({var}**3)"
    model = smf.ols(formula=formula, data=boston).fit()
    p2 = model.pvalues.get(f"I({var} ** 2)", 1)
    p3 = model.pvalues.get(f"I({var} ** 3)", 1)
    nonlinear_results[var] = (p2, p3)
    print(f"{var:>8}: p(X^2)={p2:.4g}, p(X^3)={p3:.4g}")
```

結果顯示，indus, nox, age, dis, ptratio, 以及 medv 的平方或三次項顯著（p < 0.05），
代表這些變數與犯罪率之間存在明顯的非線性關係。
例如：

nox 與 crim 呈現遞增且彎曲的關係（污染越高，犯罪率急速上升）；

dis 與 crim 呈現遞減關係但在遠距地區趨於平緩；

age 顯示中間年齡區的地區犯罪率略高（U 型趨勢）；

ptratio 與 medv 的高次項顯著，暗示教育與房價對犯罪率的影響並非線性。

其餘變數如 zn, chas, rm, rad, tax, lstat 則未顯示顯著非線性，
表示對犯罪率的影響可由線性關係充分解釋。

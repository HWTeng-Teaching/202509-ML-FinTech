Suppose that we have four observations, for which we compute a dissimilarity matrix, given by
$$
\begin{bmatrix}
   & 0.3 & 0.4 & 0.7 \\
0.3 &     & 0.5 & 0.8 \\
0.4 & 0.5 &     & 0.45 \\
0.7 & 0.8 & 0.45 & 
\end{bmatrix}
$$
For instance, the dissimilarity between the first and second observations is 0.3, and the dissimilarity between the second and fourth
observations is 0.8.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage,  dendrogram
from scipy.spatial.distance import squareform

D = np.array([
    [0,0.3,0.4,0.7],
    [0.3,0,0.5,0.8],
    [0.4,0.5,0,0.45],
    [0.7,0.8,0.45,0]
],dtype=float)
labels = np.array(['1','2','3','4'])

d_condensed = squareform(D)
```

(a) On the basis of this dissimilarity matrix, sketch the dendrogram
that results from hierarchically clustering these four observations using complete linkage. Be sure to indicate on the plot the
height at which each fusion occurs, as well as the observations
corresponding to each leaf in the dendrogram.

```python
z_complete = linkage(d_condensed, method = 'complete')

plt.figure(figsize=(7,4))
dendrogram(z_complete, labels=labels, show_leaf_counts=False)
plt.title('Complete Linkage Dendrogram')
plt.xlabel('Observations')
plt.ylabel('Height(dissimilarity)')
plt.tight_layout()
plt.show()
```

(b) Repeat (a), this time using single linkage clustering.

```python
z_single = linkage(d_condensed, method='single')

plt.figure(figsize=(7,4))
dendrogram(z_single, labels = labels, show_leaf_counts=False)
plt.title('Single Linkage Dendrogram')
plt.xlabel('Observations')
plt.ylabel('Height (dissimilarity)')
plt.tight_layout()
plt.show()
```

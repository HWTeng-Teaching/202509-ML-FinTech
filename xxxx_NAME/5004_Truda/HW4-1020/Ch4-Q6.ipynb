6. Suppose we collect data for a group of students in a statistics class with variables $X_1 = \text{hours studied}$, $X_2 = \text{undergrad GPA}$, and $Y = \text{receive an A}$. We fit a logistic regression and produce estimated coefficients, $\hat{\beta}_0 = -6, \hat{\beta}_1 = 0.05, \hat{\beta}_2 = 1$.


(a) Estimate the probability that a student who studies for $40 \text{ h}$ and has an undergrad $\text{GPA}$ of $3.5$ gets an A in the class.

```python
import numpy as np

# Estimated coefficients
beta_0 = -6.0
beta_1 = 0.05
beta_2 = 1.0

# Input values for part (a)
X1_a = 40.0  # hours studied
X2_a = 3.5   # undergrad GPA

# Function to calculate the linear predictor (eta)
def calculate_eta(b0, b1, X1, b2, X2):
    return b0 + b1 * X1 + b2 * X2

# Function to calculate the probability (P) using the logistic function
def calculate_probability(eta):
    # P = exp(eta) / (1 + exp(eta))
    return np.exp(eta) / (1 + np.exp(eta))

# --- Solution for (a) ---
eta_a = calculate_eta(beta_0, beta_1, X1_a, beta_2, X2_a)
probability_a = calculate_probability(eta_a)

print(f"--- Solution for (a) ---")
print(f"Linear Predictor (eta) for (a): {eta_a:.4f}")
print(f"Estimated Probability for (a): {probability_a:.4f}")
print(f"The student has an estimated probability of approximately {probability_a*100:.2f}% of getting an A.")
```

(b) How many hours would the student in part (a) need to study to have a $50\%$ chance of getting an A in the class?

```python
# --- Solution for (b) ---
# We need P(Y=1) = 0.5, which means the linear predictor eta must be 0.
# eta = beta_0 + beta_1 * X1_b + beta_2 * X2_b = 0

# Given values:
X2_b = 3.5  # undergrad GPA remains the same as part (a)
# We solve for X1_b:
# 0 = -6 + 0.05 * X1_b + 1 * 3.5
# 0 = -2.5 + 0.05 * X1_b
# X1_b = 2.5 / 0.05

X1_b = (-(beta_0 + beta_2 * X2_b)) / beta_1

print(f"\n--- Solution for (b) ---")
print(f"For a 50% chance, the linear predictor (eta) must be 0.")
print(f"The required hours studied (X1) for 50% chance: {X1_b:.2f} hours.")
print(f"The student would need to study for {X1_b:.2f} hours.")

# Combined execution of all Python code
if __name__ == '__main__':
    # Rerunning all the steps together for a final output
    print("\n\n--- Full Combined Python Execution Output ---")
    
    # Part (a)
    eta_a = calculate_eta(beta_0, beta_1, X1_a, beta_2, X2_a)
    probability_a = calculate_probability(eta_a)
    print(f"\n(a) Probability for X1=40, X2=3.5:")
    print(f"   eta = {eta_a:.4f}")
    print(f"   P(Y=1) = {probability_a:.4f}")

    # Part (b)
    # X1_b = 50
    X1_b = (-(beta_0 + beta_2 * X2_b)) / beta_1
    print(f"\n(b) Hours needed for 50% chance (P(Y=1)=0.5, eta=0):")
    print(f"   Required hours (X1) = {X1_b:.2f} hours")
```

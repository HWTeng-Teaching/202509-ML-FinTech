12. Suppose that you wish to classify an observation $X \in \mathbb{R}$ into $\text{apples}$ and $\text{oranges}$. You fit a logistic regression model and find that
$$
\widehat{\Pr}(Y = \text{orange}|X = x) = \frac{\exp(\hat{\beta}_0 + \hat{\beta}_1 x)}{1 + \exp(\hat{\beta}_0 + \hat{\beta}_1 x)}.
$$
Your friend fits a logistic regression model to the same data using the $\text{softmax}$ formulation in $(4.13)$, and finds that
$$
\widehat{\Pr}(Y = \text{orange}|X = x) = \frac{\exp(\hat{\alpha}_{\text{orange}0} + \hat{\alpha}_{\text{orange}1} x)}{\exp(\hat{\alpha}_{\text{orange}0} + \hat{\alpha}_{\text{orange}1} x) + \exp(\hat{\alpha}_{\text{apple}0} + \hat{\alpha}_{\text{apple}1} x)}.
$$

(a) What is the log odds of $\text{orange}$ versus $\text{apple}$ in your model?
(b) What is the log odds of $\text{orange}$ versus $\text{apple}$ in your friend's model?
(c) Suppose that in your model, $\hat{\beta}_0 = 2$ and $\hat{\beta}_1 = -1$. What are the coefficient estimates in your friend's model? Be as specific as possible.
(d) Now suppose that you and your friend fit the same two models on a different data set. This time, your friend gets the coefficient estimates $\hat{\alpha}_{\text{orange}0} = 1.2$, $\hat{\alpha}_{\text{orange}1} = -2$, $\hat{\alpha}_{\text{apple}0} = 3$, $\hat{\alpha}_{\text{apple}1} = 0.6$. What are the coefficient estimates in your model?
(e) Finally, suppose you apply both models from (d) to a data set with $2,000$ test observations. What fraction of the time do you expect the predicted class labels from your model to agree with those from your friend's model? Explain your answer.

```python
import numpy as np

# --- 12. Logistic Regression vs. Softmax Problem ---

# (a) & (b) Log Odds Concept: The code confirms the identity but doesn't calculate
# a specific number as the answer is a function of X.
print("--- Solution for (a) & (b) Log Odds ---")
print("(a) Your model Log Odds (orange vs. apple): beta_0 + beta_1 * X")
print("(b) Friend's model Log Odds (orange vs. apple): (alpha_orange0 - alpha_apple0) + (alpha_orange1 - alpha_apple1) * X")
print("These two expressions must be mathematically equal.")


# --- (c) Calculate Friend's coefficients given Your coefficients ---
print("\n--- Solution for (c) Calculation ---")
beta_0_c = 2.0
beta_1_c = -1.0

# Relationship:
# alpha_orange0 - alpha_apple0 = beta_0_c
# alpha_orange1 - alpha_apple1 = beta_1_c

print(f"Given Your coefficients: beta_0 = {beta_0_c}, beta_1 = {beta_1_c}")
print(f"Friend's coefficients must satisfy:")
print(f"  alpha_orange0 - alpha_apple0 = {beta_0_c}")
print(f"  alpha_orange1 - alpha_apple1 = {beta_1_c}")

# Example 1: Setting apple as the reference class (alpha_apple = 0)
alpha_apple0_c_ex1 = 0.0
alpha_apple1_c_ex1 = 0.0
alpha_orange0_c_ex1 = beta_0_c + alpha_apple0_c_ex1  # 2.0
alpha_orange1_c_ex1 = beta_1_c + alpha_apple1_c_ex1  # -1.0
print("\nExample 1 (Setting apple as ref class):")
print(f"  alpha_orange0 = {alpha_orange0_c_ex1}, alpha_orange1 = {alpha_orange1_c_ex1}")
print(f"  alpha_apple0 = {alpha_apple0_c_ex1}, alpha_apple1 = {alpha_apple1_c_ex1}")

# Example 2: Setting orange as the reference class (alpha_orange = 0)
alpha_orange0_c_ex2 = 0.0
alpha_orange1_c_ex2 = 0.0
alpha_apple0_c_ex2 = alpha_orange0_c_ex2 - beta_0_c  # -2.0
alpha_apple1_c_ex2 = alpha_orange1_c_ex2 - beta_1_c  # 1.0
print("Example 2 (Setting orange as ref class):")
print(f"  alpha_orange0 = {alpha_orange0_c_ex2}, alpha_orange1 = {alpha_orange1_c_ex2}")
print(f"  alpha_apple0 = {alpha_apple0_c_ex2}, alpha_apple1 = {alpha_apple1_c_ex2}")


# --- (d) Calculate Your coefficients given Friend's coefficients ---
print("\n--- Solution for (d) Calculation ---")
alpha_orange0_d = 1.2
alpha_orange1_d = -2.0
alpha_apple0_d = 3.0
alpha_apple1_d = 0.6

# Calculate Your model's coefficients
beta_0_d = alpha_orange0_d - alpha_apple0_d
beta_1_d = alpha_orange1_d - alpha_apple1_d

print(f"Friend's coefficients: a_orange0={alpha_orange0_d}, a_orange1={alpha_orange1_d}, a_apple0={alpha_apple0_d}, a_apple1={alpha_apple1_d}")
print("Your model coefficients (beta = alpha_orange - alpha_apple):")
print(f"  beta_0 = {beta_0_d:.2f}")
print(f"  beta_1 = {beta_1_d:.2f}")


# --- (e) Fraction of Agreement ---
print("\n--- Solution for (e) Calculation & Explanation ---")
test_observations = 2000
expected_agreement = 1.0 # 100%

print("Expected Fraction of Agreement: 1.0 (or 100%)")
print(f"Expected number of agreed classifications out of {test_observations} observations: {test_observations * expected_agreement:.0f}")

# The explanation is conceptual, so we output the key text.
print("\nExplanation (Conceptual):")
print("Since Binary Logistic Regression and Softmax with two classes (fitted to the same data) are mathematically equivalent models, their Log Odds functions are identical.")
print("This identity ensures that the estimated probabilities for Y=orange at any given X are identical for both models.")
print("As both models use the same prediction rule (choose the class with the highest probability, which is equivalent to Log Odds > 0), their predicted class labels will agree 100% of the time.")
```

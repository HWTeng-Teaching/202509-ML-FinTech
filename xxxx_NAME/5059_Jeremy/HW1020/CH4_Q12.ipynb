## Problem 12

Suppose that you wish to classify an observation $X \in \mathbb{R}$ into apples and oranges. You fit a logistic regression model and find that

$$\Pr(Y = \text{orange} \mid X = x) = \frac{\exp(\hat{\beta}_0 + \hat{\beta}_1 x)}{1 + \exp(\hat{\beta}_0 + \hat{\beta}_1 x)}$$

Your friend fits a logistic regression model to the same data using the softmax formulation in (4.13), and finds that

$$\Pr(Y = \text{orange} \mid X = x) = \frac{\exp(\hat{\alpha}_{\text{orange0}} + \hat{\alpha}_{\text{orange1}} x)}{\exp(\hat{\alpha}_{\text{orange0}} + \hat{\alpha}_{\text{orange1}} x) + \exp(\hat{\alpha}_{\text{apple0}} + \hat{\alpha}_{\text{apple1}} x)}$$

 (a) What is the log odds of orange versus apple in your model?

**Solution (a):**

In the logistic regression model, we have:

$$\Pr(Y = \text{orange} \mid X = x) = \frac{\exp(\hat{\beta}_0 + \hat{\beta}_1 x)}{1 + \exp(\hat{\beta}_0 + \hat{\beta}_1 x)}$$

Since this is a binary classification problem, we know that:

$$\Pr(Y = \text{apple} \mid X = x) = 1 - \Pr(Y = \text{orange} \mid X = x) = \frac{1}{1 + \exp(\hat{\beta}_0 + \hat{\beta}_1 x)}$$

The **odds** of orange versus apple is:

$$\text{odds} = \frac{\Pr(Y = \text{orange} \mid X = x)}{\Pr(Y = \text{apple} \mid X = x)} = \frac{\frac{\exp(\hat{\beta}_0 + \hat{\beta}_1 x)}{1 + \exp(\hat{\beta}_0 + \hat{\beta}_1 x)}}{\frac{1}{1 + \exp(\hat{\beta}_0 + \hat{\beta}_1 x)}} = \exp(\hat{\beta}_0 + \hat{\beta}_1 x)$$

Therefore, the **log odds** is:

$$\log(\text{odds}) = \log\left(\frac{\Pr(Y = \text{orange} \mid X = x)}{\Pr(Y = \text{apple} \mid X = x)}\right) = \hat{\beta}_0 + \hat{\beta}_1 x$$

**Answer:** The log odds of orange versus apple in the model is $\hat{\beta}_0 + \hat{\beta}_1 x$.

(b) What is the log odds of orange versus apple in your friend’s
 model?

**Solution (b):**

In the softmax formulation, your friend's model gives:

$$\Pr(Y = \text{orange} \mid X = x) = \frac{\exp(\hat{\alpha}_{\text{orange0}} + \hat{\alpha}_{\text{orange1}} x)}{\exp(\hat{\alpha}_{\text{orange0}} + \hat{\alpha}_{\text{orange1}} x) + \exp(\hat{\alpha}_{\text{apple0}} + \hat{\alpha}_{\text{apple1}} x)}$$

By the property of softmax, we can write:

$$\Pr(Y = \text{apple} \mid X = x) = \frac{\exp(\hat{\alpha}_{\text{apple0}} + \hat{\alpha}_{\text{apple1}} x)}{\exp(\hat{\alpha}_{\text{orange0}} + \hat{\alpha}_{\text{orange1}} x) + \exp(\hat{\alpha}_{\text{apple0}} + \hat{\alpha}_{\text{apple1}} x)}$$

The **odds** of orange versus apple is:

$$\text{odds} = \frac{\Pr(Y = \text{orange} \mid X = x)}{\Pr(Y = \text{apple} \mid X = x)} = \frac{\exp(\hat{\alpha}_{\text{orange0}} + \hat{\alpha}_{\text{orange1}} x)}{\exp(\hat{\alpha}_{\text{apple0}} + \hat{\alpha}_{\text{apple1}} x)}$$

Therefore, the **log odds** is:

$$\log(\text{odds}) = \log\left(\frac{\exp(\hat{\alpha}_{\text{orange0}} + \hat{\alpha}_{\text{orange1}} x)}{\exp(\hat{\alpha}_{\text{apple0}} + \hat{\alpha}_{\text{apple1}} x)}\right)$$

$$= (\hat{\alpha}_{\text{orange0}} + \hat{\alpha}_{\text{orange1}} x) - (\hat{\alpha}_{\text{apple0}} + \hat{\alpha}_{\text{apple1}} x)$$

$$= (\hat{\alpha}_{\text{orange0}} - \hat{\alpha}_{\text{apple0}}) + (\hat{\alpha}_{\text{orange1}} - \hat{\alpha}_{\text{apple1}}) x$$

**Answer:** The log odds of orange versus apple in your friend's model is:
$$(\hat{\alpha}_{\text{orange0}} - \hat{\alpha}_{\text{apple0}}) + (\hat{\alpha}_{\text{orange1}} - \hat{\alpha}_{\text{apple1}}) x$$

 (c) Suppose that in your model, ˆβ0 = 2 and ˆβ1 = −1. What are
 the coefficient estimates in your friend’s model? Be as specific
 as possible.

**Solution (c):**

From parts (a) and (b), we know that both models are fitting the same data, so they must produce the **same log odds**:

From **(a)**: Log odds = $\hat{\beta}_0 + \hat{\beta}_1 x$

From **(b)**: Log odds = $(\hat{\alpha}_{\text{orange0}} - \hat{\alpha}_{\text{apple0}}) + (\hat{\alpha}_{\text{orange1}} - \hat{\alpha}_{\text{apple1}}) x$

Since these must be equal for all values of $x$:

$$\hat{\beta}_0 + \hat{\beta}_1 x = (\hat{\alpha}_{\text{orange0}} - \hat{\alpha}_{\text{apple0}}) + (\hat{\alpha}_{\text{orange1}} - \hat{\alpha}_{\text{apple1}}) x$$

Matching coefficients:
- **Intercept**: $\hat{\beta}_0 = \hat{\alpha}_{\text{orange0}} - \hat{\alpha}_{\text{apple0}}$
- **Slope**: $\hat{\beta}_1 = \hat{\alpha}_{\text{orange1}} - \hat{\alpha}_{\text{apple1}}$

Given $\hat{\beta}_0 = 2$ and $\hat{\beta}_1 = -1$:

$$\hat{\alpha}_{\text{orange0}} - \hat{\alpha}_{\text{apple0}} = 2$$
$$\hat{\alpha}_{\text{orange1}} - \hat{\alpha}_{\text{apple1}} = -1$$

**Key insight**: In softmax regression, we have the freedom to add/subtract any constant from all class parameters without changing the predictions. This is because:

$$\frac{e^{a}}{e^{a} + e^{b}} = \frac{e^{a+c}}{e^{a+c} + e^{b+c}}$$

Therefore, we can set one class's parameters to 0 (a common convention). If we set:
$$\hat{\alpha}_{\text{apple0}} = 0 \quad \text{and} \quad \hat{\alpha}_{\text{apple1}} = 0$$

Then:
$$\hat{\alpha}_{\text{orange0}} = 2 \quad \text{and} \quad \hat{\alpha}_{\text{orange1}} = -1$$

**Answer:** 
- $\hat{\alpha}_{\text{orange0}} - \hat{\alpha}_{\text{apple0}} = 2$
- $\hat{\alpha}_{\text{orange1}} - \hat{\alpha}_{\text{apple1}} = -1$

More specifically, one valid solution is:
- $\hat{\alpha}_{\text{orange0}} = 2, \hat{\alpha}_{\text{orange1}} = -1$
- $\hat{\alpha}_{\text{apple0}} = 0, \hat{\alpha}_{\text{apple1}} = 0$

However, any values satisfying the difference constraints above are valid (the model is **not identifiable** without additional constraints).

 (d) Now suppose that you and your friend fit the same two models
 on a different data set. This time, your friend gets the coefficient
 estimates ˆαorange0 = 1.2, ˆαorange1 = −2, ˆαorange0 = 3, ˆαorange1 =
 0.6. What are the coefficient estimates in your model?

**Solution (d):**

**Note:** The problem statement appears to have a typo. I'll assume the coefficients are:
- $\hat{\alpha}_{\text{orange0}} = 1.2, \hat{\alpha}_{\text{orange1}} = -2$
- $\hat{\alpha}_{\text{apple0}} = 3, \hat{\alpha}_{\text{apple1}} = 0.6$

From part (c), we established the relationship between the two models:

$$\hat{\beta}_0 = \hat{\alpha}_{\text{orange0}} - \hat{\alpha}_{\text{apple0}}$$
$$\hat{\beta}_1 = \hat{\alpha}_{\text{orange1}} - \hat{\alpha}_{\text{apple1}}$$

Now we can calculate:

**For the intercept:**
$$\hat{\beta}_0 = \hat{\alpha}_{\text{orange0}} - \hat{\alpha}_{\text{apple0}} = 1.2 - 3 = -1.8$$

**For the slope:**
$$\hat{\beta}_1 = \hat{\alpha}_{\text{orange1}} - \hat{\alpha}_{\text{apple1}} = -2 - 0.6 = -2.6$$

**Verification:**
The log odds in both models should be the same:
- Your model: $\hat{\beta}_0 + \hat{\beta}_1 x = -1.8 - 2.6x$
- Friend's model: $(\hat{\alpha}_{\text{orange0}} - \hat{\alpha}_{\text{apple0}}) + (\hat{\alpha}_{\text{orange1}} - \hat{\alpha}_{\text{apple1}})x = (1.2-3) + (-2-0.6)x = -1.8 - 2.6x$ ✓

**Answer:** In your model, the coefficient estimates are:
- $\hat{\beta}_0 = -1.8$
- $\hat{\beta}_1 = -2.6$

 (e) Finally, suppose you apply both models from (d) to a data set
 with 2,000 test observations. What fraction of the time do you
 expect the predicted class labels from your model to agree with
 those from your friend’s model? Explain your answer.

**Solution (e):**

The predicted class labels from the two models will agree **100% of the time** (i.e., a fraction of **1.0** or **all 2,000 observations**).

**Explanation:**

Both models make predictions based on the same decision rule - they classify an observation as "orange" if:

$$\Pr(Y = \text{orange} \mid X = x) > 0.5$$

which is equivalent to:

$$\Pr(Y = \text{orange} \mid X = x) > \Pr(Y = \text{apple} \mid X = x)$$

This happens when the **log odds > 0**:

$$\log\left(\frac{\Pr(Y = \text{orange} \mid X = x)}{\Pr(Y = \text{apple} \mid X = x)}\right) > 0$$

**For your model (from part d):**
$$\text{Log odds} = \hat{\beta}_0 + \hat{\beta}_1 x = -1.8 - 2.6x$$

Classify as orange when: $-1.8 - 2.6x > 0$

**For your friend's model (from part d):**
$$\text{Log odds} = (\hat{\alpha}_{\text{orange0}} - \hat{\alpha}_{\text{apple0}}) + (\hat{\alpha}_{\text{orange1}} - \hat{\alpha}_{\text{apple1}})x$$
$$= (1.2 - 3) + (-2 - 0.6)x = -1.8 - 2.6x$$

Classify as orange when: $-1.8 - 2.6x > 0$

**Conclusion:**

Since both models have **exactly the same log odds** ($-1.8 - 2.6x$), they will make the same classification decision for every observation. The decision boundary is the same:

$$-1.8 - 2.6x = 0 \implies x = -\frac{1.8}{2.6} \approx -0.692$$

- Both classify as **orange** when $x < -0.692$
- Both classify as **apple** when $x > -0.692$

**Answer:** The predicted class labels will agree **100% of the time** (fraction = 1.0) because the two models are mathematically equivalent - they produce identical probabilities and thus identical predictions for any given input.

6. Suppose we collect data for a group of students in a statistics class
 with variables X1 =hours studied, X2=undergrad GPA, and Y =
 receive an A. We fit a logistic regression and produce estimated
 coefficient, ˆβ0 = −6, ˆβ1 = 0.05, ˆβ2 = 1.

(a) Estimate the probability that a student who studies for 40h and
 has an undergrad GPA of 3.5 gets an A in the class.

**Solution:**

The logistic regression probability formula is:

$$P(Y=1|X) = \frac{e^{\beta_0 + \beta_1 X_1 + \beta_2 X_2}}{1 + e^{\beta_0 + \beta_1 X_1 + \beta_2 X_2}}$$

Given conditions:
- $\hat{\beta_0} = -6$
- $\hat{\beta_1} = 0.05$
- $\hat{\beta_2} = 1$
- $X_1 = 40$ (hours studied)
- $X_2 = 3.5$ (GPA)

**Calculation steps:**

1. Calculate the linear combination:
   $$z = \hat{\beta_0} + \hat{\beta_1} X_1 + \hat{\beta_2} X_2$$
   $$z = -6 + 0.05(40) + 1(3.5)$$
   $$z = -6 + 2 + 3.5 = -0.5$$

2. Calculate the probability:
   $$P(Y=1|X) = \frac{e^{-0.5}}{1 + e^{-0.5}} = \frac{0.6065}{1.6065} = 0.3775$$

**Answer:** The probability that this student will receive an A is approximately **0.3775** or **37.75%**


 (b) How many hours would the student in part (a) need to study to
 have a 50% chance of getting an A in the class?

**Solution:**

When the probability is 50%, we have:

$$P(Y=1|X) = 0.5 = \frac{e^z}{1 + e^z}$$

This simplifies to: $e^z = 1$, which means $z = 0$

Given conditions:
- $\hat{\beta_0} = -6$
- $\hat{\beta_1} = 0.05$
- $\hat{\beta_2} = 1$
- $X_2 = 3.5$ (GPA remains the same)
- $X_1 = ?$ (hours to study, what we need to find)

**Setting up the equation:**

$$z = \hat{\beta_0} + \hat{\beta_1} X_1 + \hat{\beta_2} X_2 = 0$$

$$-6 + 0.05 X_1 + 1(3.5) = 0$$

$$-6 + 0.05 X_1 + 3.5 = 0$$

$$0.05 X_1 - 2.5 = 0$$

$$0.05 X_1 = 2.5$$

$$X_1 = \frac{2.5}{0.05} = 50$$

**Answer:** The student would need to study for **50 hours** to have a 50% chance of getting an A in the class.

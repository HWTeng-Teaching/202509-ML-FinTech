 9. This problem has to do with odds.

 (a) On average, what fraction of people with an odds of 0.37 of
 defaulting on their credit card payment will in fact default?

**Solution (a):**

The relationship between odds and probability is:

$$\text{odds} = \frac{p}{1-p}$$

where $p$ is the probability of default.

Given odds = 0.37, we need to solve for $p$:

$$0.37 = \frac{p}{1-p}$$

Solving this equation:
$$0.37(1-p) = p$$
$$0.37 - 0.37p = p$$
$$0.37 = p + 0.37p$$
$$0.37 = 1.37p$$
$$p = \frac{0.37}{1.37} = \frac{0.37}{1.37} \approx 0.2701$$

**Answer:** On average, approximately **0.27** (or **27.01%**) of people with odds of 0.37 will default on their credit card payment.

 (b) Suppose that an individual has a 16% chance of defaulting on
 her credit card payment. What are the odds that she will de
fault?


**Solution (b):**

Given that the probability of default is $p = 0.16$ (16%), we need to calculate the odds using the formula:

$$\text{odds} = \frac{p}{1-p}$$

Substituting $p = 0.16$:

$$\text{odds} = \frac{0.16}{1-0.16} = \frac{0.16}{0.84} \approx 0.1905$$

**Answer:** The odds that she will default are approximately **0.19** (or more precisely, **0.1905**).

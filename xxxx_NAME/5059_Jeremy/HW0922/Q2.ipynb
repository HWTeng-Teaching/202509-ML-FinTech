```python
import numpy as np
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt
```

(a) Based on the following dissimilarity matrix, sketch the dendrogram resulting from hierarchical clustering of these four observations using complete linkage. Be sure to indicate on the plot the height at which each fusion occurs, as well as the observations corresponding to each leaf in the dendrogram.

```python
dissimilarity_matrix = np.array([
    [0.0, 0.3, 0.4, 0.7],
    [0.3, 0.0, 0.5, 0.8],
    [0.4, 0.5, 0.0, 0.45],
    [0.7, 0.8, 0.45, 0.0]
])

# Convert the upper triangular part of the dissimilarity matrix to a condensed matrix format
condensed_matrix = sch.distance.squareform(dissimilarity_matrix)

# Perform complete linkage hierarchical clustering
Z_complete = sch.linkage(condensed_matrix, method='complete')

# Plot the dendrogram
plt.figure(figsize=(8, 6))
dendrogram = sch.dendrogram(Z_complete, labels=[1, 2, 3, 4])
plt.title('Complete Linkage Dendrogram')
plt.xlabel('Observation')
plt.ylabel('Dissimilarity')
plt.show()
```

 (b) Repeat (a), this time using single linkage clustering.

```python
# Hierarchical clustering using single linkage
Z_single = sch.linkage(sch.distance.squareform(dissimilarity_matrix), method='single')

# Plot the dendrogram for single linkage
plt.figure(figsize=(8, 5))
sch.dendrogram(Z_single, labels=[1, 2, 3, 4], above_threshold_color='blue')
plt.title("Dendrogram - Single Linkage")
plt.xlabel("Observations")
plt.ylabel("Distance")
plt.show()
```

(e)
It is mentioned in this chapter that at each fusion in the dendrogram, the position of the two clusters being fused can be swapped without changing the meaning of the dendrogram. Draw a dendrogram that is equivalent to the dendrogram in (a), for which two or more of the leaves are repositioned, but for which the meaning of the dendrogram is the same.

```python
# Demonstrate that swapping leaf labels is a visual change only (use Z_complete from the dissimilarity matrix)
plt.figure(figsize=(10, 5))
sch.dendrogram(Z_complete, labels=['1', '2', '3', '4'])
plt.title('Original Dendrogram')
plt.show()

# Rearranged labels (visual only)
plt.figure(figsize=(10, 5))
sch.dendrogram(Z_complete, labels=['2', '1', '4', '3'])
plt.title('Rearranged Dendrogram (labels swapped)')
plt.show()

print('Note: Changing the labels shown on the x-axis is a visual rearrangement only; the linkage structure in Z_complete remains the same.')
```

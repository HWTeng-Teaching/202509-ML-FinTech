### (b) 
Create a new function, Power2(), that allows you to pass any two numbers, x and a, and prints out the value of x**a. You can do this by beginning your function with the line <br>
**def Power2(x, a):** <br>
You should be able to call your function by entering, for instance, <br>
**Power2(3, 8)** <br>
on the command line. This should output the value of 38, namely, 6, 561.

```python
# (b)
def Power2(x, a):
    print(f"{x}^{a} =", x ** a)

# Example
Power2(3, 8)
```

### (d) 
Now create a new function, Power3(), that actually returns the
result x\**a as a Python object, rather than simply printing it
to the screen. That is, if you store the value x**a in an object
called result within your function, then you can simply return this result, using the following line: <br>
**return result**<br>
Note that the line above should be the last line in your function, and it should be indented 4 spaces.

```python
# (d)
def Power3(x, a):
    result = x ** a
    return result

# Example
value = Power3(2, 5)
print("2^5 returned value =", value)
```

### (e)
Now using the Power3() function, create a plot of f(x) = x2. The x-axis should display a range of integers from 1 to 10, and the y-axis should display x2. Label the axes appropriately, and use an appropriate title for the figure. Consider displaying either the x-axis, the y-axis, or both on the log-scale. You can do this by using the ax.set_xscale() and ax.set_yscale() methods of the axes you are plotting to.

```python
# (e)
import numpy as np
import matplotlib.pyplot as plt

# 使用 Power3 計算 y = x^2
x = np.arange(1, 11)
y = [Power3(i, 2) for i in x]

fig, ax = plt.subplots(figsize=(6,4))
ax.plot(x, y, marker='o', linestyle='-', color='steelblue')
ax.set_xlabel('x')
ax.set_ylabel('$x^2$')
ax.set_title('Plot of f(x) = $x^2$')

# 可選：設定 y 軸為對數刻度
ax.set_yscale('log')
plt.show()
```

### (f) 
Create a function, PlotPower(), that allows you to create a plot of x against x**a for a fixed a and a sequence of values of x. For instance, if you call <br>
**PlotPower(np.arange(1, 11), 3)** <br>
then a plot should be created with an x-axis taking on values 1,2,...,10, and a y-axis taking on values $1^3$, $2^3$, ...,$10^3$.


```python
# (f)
def PlotPower(x, a):
    y = [Power3(i, a) for i in x]
    fig, ax = plt.subplots(figsize=(6,4))
    ax.plot(x, y, marker='o', color='darkorange')
    ax.set_xlabel('x')
    ax.set_ylabel(f'$x^{a}$')
    ax.set_title(f'Plot of f(x) = $x^{a}$')
    plt.show()

# Example: a = 3, x = 1~10
PlotPower(np.arange(1, 11), 3)
```

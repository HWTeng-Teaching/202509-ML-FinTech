We will now consider the Boston housing data set, from the ISLP library.

```python
import numpy as np
import pandas as pd
from ISLP import load_data
import matplotlib.pyplot as plt
```

```python
# Load Boston dataset
Boston = load_data('Boston')
Boston.head()
```

**(a)** Estimate the population mean of `medv`.

```python
mu_hat = Boston['medv'].mean()
mu_hat
```

**(b)** Estimate the standard error of the mean of `medv`.

```python
n = len(Boston)
se_hat = Boston['medv'].std(ddof=1) / np.sqrt(n)
se_hat
```

**(c)** Bootstrap estimate of the standard error of `mu_hat`.

```python
B = 5000
boot_means = []

rng = np.random.default_rng(0)

for _ in range(B):
    sample = Boston['medv'].sample(n, replace=True, random_state=rng.integers(1e9))
    boot_means.append(sample.mean())

boot_se = np.std(boot_means, ddof=1)
boot_se
```

**(d)** Bootstrap 95% CI and comparison with 2SE rule.

```python
# Bootstrap percentile CI
ci_boot = np.percentile(boot_means, [2.5, 97.5])

# Two-standard-error rule
ci_2se = [mu_hat - 2*se_hat, mu_hat + 2*se_hat]

ci_boot, ci_2se
```

**(e)** Estimate the population median of `medv`.

```python
mu_med = Boston['medv'].median()
mu_med
```

**(f)** Bootstrap estimate of SE for the median of `medv`.

```python
boot_medians = []

rng = np.random.default_rng(1)

for _ in range(B):
    sample = Boston['medv'].sample(n, replace=True, random_state=rng.integers(1e9))
    boot_medians.append(sample.median())

boot_se_median = np.std(boot_medians, ddof=1)
boot_se_median
```

**(g)** Estimate the 10th percentile of `medv`.

```python
mu_0_1 = np.percentile(Boston['medv'], 10)
mu_0_1
```

**(h)** Bootstrap estimate of SE for the 10th percentile.

```python
boot_pct10 = []

rng = np.random.default_rng(2)

for _ in range(B):
    sample = Boston['medv'].sample(n, replace=True, random_state=rng.integers(1e9))
    boot_pct10.append(np.percentile(sample, 10))

boot_se_pct10 = np.std(boot_pct10, ddof=1)
boot_se_pct10
```

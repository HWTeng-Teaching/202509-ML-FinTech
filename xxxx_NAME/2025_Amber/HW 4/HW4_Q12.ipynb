Suppose you want to classify an observation X ∈ R into two classes: apples and oranges.

You fit a logistic regression model and find:
P(Y = orange | X = x) = exp(b0 + b1 * x) / (1 + exp(b0 + b1 * x))

Your friend fits a logistic regression model to the same data using a two-class softmax and finds:
P(Y = orange | X = x) =
  exp(a_orange0 + a_orange1 * x)
  / [ exp(a_orange0 + a_orange1 * x) + exp(a_apple0 + a_apple1 * x) ]

We compare two equivalent parameterizations of binary classification with one feature x.

---

### Your Model (Logistic)

P(Y = orange | x) = exp(b0 + b1x) / (1 + exp(b0 + b1x))

---

### Friend’s Model (Two-class Softmax)

P(Y = orange | x) =
exp(a_orange0 + a_orange1x) /
[ exp(a_orange0 + a_orange1x) + exp(a_apple0 + a_apple1x) ]

(a) What is the log-odds of orange versus apple in your model?


log[p / (1 - p)] = b0 + b1x

(b) What is the log-odds of orange versus apple in your friend’s model?


log[p / (1 - p)] = (a_orange0 - a_apple0) + (a_orange1 - a_apple1)x

(c) Suppose that in your model, b0 = 2 and b1 = -1.
    What are the coefficient values in your friend’s model? Be as specific as possible.

Any set satisfying  
(a_orange0 - a_apple0) = 2  
(a_orange1 - a_apple1) = -1

Example (set apple as baseline):
a_apple0 = 0, a_apple1 = 0  
a_orange0 = 2, a_orange1 = -1

Example above; infinite valid sets exist.

Only the *difference* between coefficients affects predictions; adding a constant to both class scores changes nothing.

(d) Now suppose that you and your friend fit the same two models on a different data set.
    This time, your friend gets the estimates:
    a_orange0 = 1.2, a_orange1 = -2, a_apple0 = 3, a_apple1 = 0.6.
    What are the coefficient values in your model?

a_orange0 = 1.2, a_orange1 = -2  
a_apple0 = 3, a_apple1 = 0.6

Then:
b0 = a_orange0 - a_apple0 = 1.2 - 3 = -1.8  
b1 = a_orange1 - a_apple1 = -2 - 0.6 = -2.6


(e) Finally, suppose you apply both models from (d) to a data set with 2,000 test observations.
    What fraction of the time do you expect the predicted class labels from your model to agree
    with those from your friend’s model? Explain your answer.

Because both models yield identical probabilities (since log-odds depend only on parameter differences), their predicted labels are identical.

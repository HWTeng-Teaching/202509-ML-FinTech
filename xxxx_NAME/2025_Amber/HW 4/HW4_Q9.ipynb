This problem has to do with *odds*.

**(a)** On average, what fraction of people with an odds of 0.37 of defaulting on their credit card payment will in fact default?  


- From odds to probability:  p = odds / (1 + odds)
- From probability to odds: odds = p / (1 - p)

p = 0.37 / (1 + 0.37) = 0.37 / 1.37 ≈ 0.2701

**(b)** Suppose that an individual has a 16% chance of defaulting on her credit card payment.  
What are the odds that she will default?

odds = 0.16 / (1 - 0.16) = 0.16 / 0.84 ≈ 0.1905

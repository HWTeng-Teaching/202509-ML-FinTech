Suppose that you wish to classify an observation $X \in \mathbb{R}$ into $\text{apples}$ and $\text{oranges}$. You fit a logistic regression model and find that
$$
\hat{\Pr}(Y = \text{orange}|X = x) = \frac{\exp(\hat{\beta}_0 + \hat{\beta}_1 x)}{1 + \exp(\hat{\beta}_0 + \hat{\beta}_1 x)}.
$$
Your friend fits a logistic regression model to the same data using the $\text{softmax}$ formulation in (4.13), and finds that
$$
\hat{\Pr}(Y = \text{orange}|X = x) = \frac{\exp(\hat{\alpha}_{\text{orange}0} + \hat{\alpha}_{\text{orange}1}x)}{\exp(\hat{\alpha}_{\text{orange}0} + \hat{\alpha}_{\text{orange}1}x) + \exp(\hat{\alpha}_{\text{apple}0} + \hat{\alpha}_{\text{apple}1}x)}.
$$

(a) What is the log odds of orange versus apple in your model?

y = exp(beta0 + beta1 * x)

log odds 

= ln(Pr(Y = orange) / Pr(Y = apple)) 

= ln(Pr(Y = orange) / (1 - Pr(Y = orange))) 

= ln((y / (1 + y)) / (1 - (y / (1 + y))))

= ln((y / (1 + y)) / (1 / (1 + y)))

= ln(y)

= ln(exp(beta0 + beta1 * x))

= beta0 + beta1 * x

(b) What is the log odds of orange versus apple in your friend’s
model?

y1 = exp(alpha_orange0 + alpha_orange1 * x)

y2 = exp(alpha_apple0 + alpha_apple1 * x)

log odds

= ln(Pr(Y = orange) / Pr(Y = apple))

= ln(Pr(Y = orange) / (1 - Pr(Y = orange))) 

= ln((y1 / (y1 + y2)) / (1 - (y1 / (y1 + y2))))

= ln((y1 / (y1 + y2)) / (y2 / (y1 + y2)))

= ln(y1 / y2)

= ln(y1) - ln(y2)

= ln(exp(alpha_orange0 + alpha_orange1 * x)) - ln(exp(alpha_apple0 + alpha_apple1 * x))

= (alpha_orange0 + alpha_orange1 * x) - (alpha_apple0 + alpha_apple1 * x)

= (alpha_orange0 - alpha_apple0) + (alpha_orange1 - alpha_apple1) * x

(c) Suppose that in your model, $\hat{\beta}_0 = 2$ and $\hat{\beta}_1 = -1$. What are the coefficient estimates in your friend’s model? Be as specific as possible.

since the log odds of two model should be equal

-> beta0 + beta1 * x = (alpha_orange0 - alpha_apple0) + (alpha_orange1 - alpha_apple1) * x

-> beta0 = alpha_orange0 - alpha_apple0, beta1 = alpha_orange1 - alpha_apple1

and beta0 = 2, beta1 = -1

-> alpha_orange0 - alpha_apple0 = 2,  alpha_orange1 - alpha_apple1 = -1

The exact value of alpha_apple0 can only be deternined once alpha_orange0 is set, and vice versa. Similarly, the value of alpha_apple1 is dependent on alpha_orange1 due to the lack of parameter identifiability in the Softmax model.





(d) Now suppose that you and your friend fit the same two models on a different data set. This time, your friend gets the coefficient estimates $\hat{\alpha}_{\text{orange}0} = 1.2$, $\hat{\alpha}_{\text{orange}1} = -2$, $\hat{\alpha}_{\text{apple}0} = 3$, $\hat{\alpha}_{\text{apple}1} = 0.6$. What are the coefficient estimates in your model?

beta0 

= alpha_orange0 - alpha_apple0 
   
= 1.2 - 3

= -1.8


beta1 

= alpha_orange1 - alpha_apple1

= -2 - 0.6

= -2.6

(e) Finally, suppose you apply both models from (d) to a data set
with 2,000 test observations. What fraction of the time do you
expect the predicted class labels from your model to agree with
those from your friend’s model? Explain your answer.

In binary logic regression, once two model have the same log odds, the classification outcome of two model will be the same. So the predicted class labels of my model will always agree with my friend's model.

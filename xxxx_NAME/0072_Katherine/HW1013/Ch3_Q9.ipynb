(c)Use the sm.OLS() function to perform a multiple linear regression
with mpg as the response and all other variables except name as
the predictors. Use the summarize() function to print the results.
Comment on the output.

```python
df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')
df = df.dropna(subset=['horsepower'])
X = df.drop(columns=['mpg', 'name'])
y = df['mpg']

X = sm.add_constant(X)
model = sm.OLS(y, X).fit()

print(model.summary())
```

i. Is there a relationship between the predictors and the response? Use the anova_lm() function from statsmodels to
answer this question.

```python
formula = 'mpg ~ cylinders + displacement + horsepower + weight + acceleration + year + origin'
model_formula = smf.ols(formula, data=df).fit()
anova_results = anova_lm(model_formula)
print(anova_results)
```

Based on the p-values, there is strong evidence that predictors including cylinders, displacement, horsepower, weight, year, and origin have a statistically significant relationship with the response variable mpg. Only acceleration does not show a significant association with mpg.

ii. Which predictors appear to have a statistically significant
relationship to the response?

```python
significant_predictors = model.pvalues[model.pvalues < 0.05].index.tolist()
# 移除截距
significant_predictors = [var for var in significant_predictors if var != 'const']
print("statistically significant predictor:", significant_predictors)
```

iii. What does the coefficient for the year variable suggest?

It represents the expected change in mpg for a one-unit increase in year increases, holding all other variables constant. 

(d) Produce some of diagnostic plots of the linear regression fit as
described in the lab. Comment on any problems you see with the
fit. Do the residual plots suggest any unusually large outliers?
Does the leverage plot identify any observations with unusually
high leverage?

```python
# 取得模型殘差與預測值
fitted_vals = model.fittedvalues
residuals = model.resid
standardized_resid = model.get_influence().resid_studentized_internal

# Residuals vs Fitted
plt.figure(figsize=(6,4))
plt.scatter(fitted_vals, residuals, alpha=0.7)
plt.axhline(0, color='red', linestyle='--')
plt.xlabel('Fitted values')
plt.ylabel('Residuals')
plt.title('Residuals vs Fitted')
plt.show()

# QQ plot
sm.qqplot(residuals, line='45', fit=True)
plt.title('Normal Q-Q Plot')
plt.show()

# Scale-Location plot (sqrt of standardized residuals vs fitted)
plt.figure(figsize=(6,4))
plt.scatter(fitted_vals, abs(standardized_resid)**0.5, alpha=0.7)
plt.xlabel('Fitted values')
plt.ylabel('Sqrt(|Standardized residuals|)')
plt.title('Scale-Location Plot')
plt.show()

# Leverage plot
from statsmodels.graphics.regressionplots import influence_plot
plt.figure(figsize=(8,6))
influence_plot(model, criterion="cooks")
plt.title('Leverage / Influence Plot')
plt.show()
```

The diagnostic plots suggest the following about the linear regression fit:

The Residuals vs Fitted plot shows no clear pattern, though some points are far from zero, indicating potential outliers.

The Normal Q-Q plot shows that most residuals lie near the diagonal line, suggesting approximate normality.

The Scale-Location plot indicates roughly constant variance, as no obvious pattern is observed.

The Leverage and Influence plot identifies some observations with unusually high leverage, which may be influential points or outliers.

In summary, while the model appears generally reasonable, there is evidence of a few potential outliers and influential observations that could affect the fit.

(e) Fit some models with interactions as described in the lab. Do any interactions appear to be statistically significant?

```python
# 數值型 predictor
X = df.drop(columns=['mpg', 'name']).astype(float)
y = df['mpg'].astype(float)

# 加入交互作用: horsepower * weight
X['horsepower_weight'] = X['horsepower'] * X['weight']

# 可以加入多個交互作用，例如:
# X['displacement_acceleration'] = X['displacement'] * X['acceleration']

# 加截距
X_const = sm.add_constant(X)

# 多元回歸
model_interact = sm.OLS(y, X_const).fit()

# 顯示摘要
print(model_interact.summary())
```

```python
from itertools import combinations

X_base = df.drop(columns=['mpg', 'name']).astype(float)
y = df['mpg'].astype(float)

# 生成兩兩交互作用
for a, b in combinations(X_base.columns, 2):
    X_base[f'{a}_x_{b}'] = X_base[a] * X_base[b]

# 加截距
X_const = sm.add_constant(X_base)

# 拟合
model_all_interact = sm.OLS(y, X_const).fit()

# 顯示摘要
print(model_all_interact.summary())

# 列出顯著交互作用
significant_interactions = model_all_interact.pvalues[model_all_interact.pvalues < 0.05].index.tolist()
# 移除截距
significant_interactions = [var for var in significant_interactions if var != 'const']
print("顯著交互作用:", significant_interactions)
```

Fitting a multiple linear regression model including all pairwise interactions, the statistically significant effects are displacement, acceleration, origin, and the interactions displacement × year, acceleration × year, acceleration × origin. This indicates that the effect of some predictors on mpg depends on other variables. Other interactions are not statistically significant.

(f) Try a few different transformations of the variables, such as
log(X), √X, X2. Comment on your findings.

```python
# 去掉非數值欄位
X_orig = df.drop(columns=['mpg', 'name']).astype(float)
y = df['mpg'].astype(float)

# 範例：對 'horsepower' 做不同轉換
transform_results = {}

for transform in ['none', 'log', 'sqrt', 'square']:
    X = X_orig.copy()
    
    if transform == 'log':
        # 避免 log(0)，加一個小常數
        X['horsepower'] = np.log(X['horsepower'] + 1)
    elif transform == 'sqrt':
        X['horsepower'] = np.sqrt(X['horsepower'])
    elif transform == 'square':
        X['horsepower'] = X['horsepower'] ** 2
    # 'none' 則保持原樣

    # 加截距
    X_const = sm.add_constant(X)
    
    # 拟合回歸
    model = sm.OLS(y, X_const).fit()
    
    # 儲存結果
    transform_results[transform] = model.rsquared_adj  # 或使用其他指標，如 AIC、BIC
    print(f"Transformation: {transform}")
    print(model.summary())
    print("-"*60)
    
print("Adjusted R-squared for different transformations:")
print(transform_results)
```

The p-values for the model after each transformation are as follows:

none : 2.04e-139 

log : 7.43e-147 

square root : 5.83e-142 

square : 6.63e-140 

These results indicate that the log transformation produces the most statistically significant model, followed by the square root transformation, and then the square transformation.

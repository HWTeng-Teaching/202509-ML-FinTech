We will now consider the Boston housing data set, from the ISLP
library.

(a) Based on this data set, provide an estimate for the population
mean of medv. Call this estimate $\hat{\mu}$

```python
df = ISLP.load_data('Boston')
print(df.shape)
print(df.describe())
mu_hat = df['medv'].mean()
print(mu_hat)
```

(b) Provide an estimate of the standard error of $\hat{\mu}$. Interpret this
result.
Hint: We can compute the standard error of the sample mean by
dividing the sample standard deviation by the square root of the
number of observations.

```python
n = df.shape[0]
s = df['medv'].std()
std_err = s / np.sqrt(n)
print(std_err)
```

The standard error of $\hat{\mu}$ is approximately $0.4085$ (in thousands of dollars). This value is the estimated standard deviation of the sample mean, quantifying the average amount the sample mean would vary from the true population mean if the sampling process were repeated many times. Since the value is small, it suggests that our estimate $\hat{\mu}$ is highly precise.

(c) Now estimate the standard error of $\hat{\mu}$ using the bootstrap. How
does this compare to your answer from (b)?

```python
B = 1000000
np.random.seed(42)
mu_hat_bootstrap_samples = []
medv_data = df['medv'].values

for i in range(B):
    # 使用 np.random.choice 進行有放回的隨機抽樣
    # replace=True 執行有放回抽樣
    bootstrap_sample = np.random.choice(medv_data, size=n, replace=True)
    
    # 計算並儲存 bootstrap 樣本的平均數
    mu_hat_bootstrap_samples.append(np.mean(bootstrap_sample))

# 計算 Bootstrap 樣本平均數的標準差，即為 SE($\hat{\mu}$)
SE_mu_hat_bootstrap = np.std(mu_hat_bootstrap_samples, ddof=1) # 使用 ddof=1 計算樣本標準差

print(SE_mu_hat_bootstrap)
```

When the number of samples are bigger, the outcome is more close to the answer from (b). 

(d) Based on your bootstrap estimate from (c), provide a 95 % confidence interval for the mean of medv. Compare it to the results
obtained by using Boston['medv'].std() and the two standard error rule (3.9).
Hint: You can approximate a 95 % confidence interval using the
formula [$\hat{\mu}$− 2SE($\hat{\mu}$),$\hat{\mu}$ + 2SE($\hat{\mu}$)].

```python
boost_start = mu_hat - 2 * SE_mu_hat_bootstrap
boost_end = mu_hat + 2 * SE_mu_hat_bootstrap

std_start = mu_hat - 2 * std_err
std_end = mu_hat + 2 * std_err

print(f"boostrap: [ {boost_start}, {boost_end} ]")
print(f"std: [ {std_start}, {std_end} ]")
```

Two interval are almost the same, the high degree of agreement between the two methods confirms the reliability of the standard error estimate. For a large sample size ($n=506$), the Central Limit Theorem holds, making the simple formula (Std) a highly accurate approximation of the sampling distribution's variability.

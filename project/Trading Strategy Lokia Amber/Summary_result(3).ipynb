```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load your results file
df = pd.read_csv('/Users/ambervo/Library/CloudStorage/OneDrive-Personal/Documents/GMBA 3/FinTech/Group Project/Trading Strategy Lokia Amber/Testing Result(3).csv')

# Preview
print(df.head())
```

```python
#RMSE Plot: Identify the most accurate forecasting model (lower error)
plt.figure(figsize=(12,6))
sns.barplot(
    data=df,
    x="Model",
    y="RMSE_mean",
    hue="FeatureSelection"
)
plt.title("RMSE Comparison Across Models & Feature Selection Methods")
plt.ylabel("Mean RMSE (Lower = Better)")
plt.xlabel("Model")
plt.legend(title="Feature Selection")
plt.show()
```

```python
#R² Plot: Identify which model captures volatility movements (higher dynamic explanatory power).
plt.figure(figsize=(12,6))
sns.barplot(
    data=df,
    x="Model",
    y="R2_mean",
    hue="FeatureSelection"
)
plt.title("R² Comparison Across Models & Feature Selection Methods")
plt.axhline(0, color="red", linestyle="--")
plt.ylabel("Mean R² (Higher = Better)")
plt.xlabel("Model")
plt.legend(title="Feature Selection")
plt.show()
```

```python
#Correlation Plot: Evaluate how well predictions track the direction of actual volatility.
plt.figure(figsize=(12,6))
sns.barplot(
    data=df,
    x="Model",
    y="Corr_mean",
    hue="FeatureSelection"
)
plt.title("Correlation Between Actual & Predicted Volatility")
plt.ylabel("Correlation (Higher = Better)")
plt.xlabel("Model")
plt.legend(title="Feature Selection")
plt.show()
```

```python
#Bubble Plot (RMSE vs Corr): Visualize the trade-off between accuracy and ability to capture dynamics; shows overall best model.
plt.figure(figsize=(12,7))
sns.scatterplot(
    data=df,
    x="RMSE_mean",
    y="Corr_mean",
    size="R2_mean",
    hue="Model",
    style="FeatureSelection",
    sizes=(50, 300)
)
plt.title("Model Comparison: RMSE vs Correlation (Bubble = R²)")
plt.xlabel("RMSE (Lower = Better)")
plt.ylabel("Correlation (Higher = Better)")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.show()
```
